# 오류 및 디버깅

코드 개발 및 데이터 분석에는 항상 약간의 시행착오가 필요하며 IPython에는 이 프로세스를 간소화하는 도구가 포함되어 있습니다.
이 섹션에서는 파이썬의 예외 보고를 제어하기 위한 몇 가지 옵션을 간략하게 다룬 다음 코드의 오류를 디버깅하기 위한 도구를 살펴봅니다.

## 예외 제어: %xmode

파이썬 스크립트가 실패하면 대부분 예외가 발생합니다.
인터프리터가 이러한 예외 중 하나를 만나면 오류 원인에 대한 정보는 *트레이스백*에서 찾을 수 있으며 파이썬 내에서 액세스할 수 있습니다.
`%xmode` 매직 함수를 사용하면 IPython에서 예외가 발생할 때 인쇄되는 정보의 양을 제어할 수 있습니다.
다음 코드를 고려하십시오.

```python
def func1(a, b):
    return a / b

def func2(x):
    a = x
    b = x - 1
    return func1(a, b)
```

```python
func2(1)
```

`func2`를 호출하면 오류가 발생하고 인쇄된 추적을 읽으면 정확히 무슨 일이 일어났는지 알 수 있습니다.
기본 모드에서 이 추적에는 오류로 이어진 각 단계의 컨텍스트를 보여주는 여러 줄이 포함됩니다.
`%xmode` 매직 함수(*예외 모드*의 약자)를 사용하면 인쇄되는 정보를 변경할 수 있습니다.

`%xmode`는 모드라는 단일 인수를 사용하며 `Plain`, `Context` 및 `Verbose`의 세 가지 가능성이 있습니다.
기본값은 방금 표시된 것과 같은 출력을 제공하는 `Context`입니다.
`Plain`은 더 간결하고 정보를 덜 제공합니다.

```python
%xmode Plain
```

```python
func2(1)
```

`Verbose` 모드는 호출되는 모든 함수의 인수를 포함하여 몇 가지 추가 정보를 추가합니다.

```python
%xmode Verbose
```

```python
func2(1)
```

이 추가 정보는 예외가 발생하는 이유를 좁히는 데 도움이 될 수 있습니다.
그렇다면 항상 `Verbose` 모드를 사용하지 않는 이유는 무엇입니까?
코드가 복잡해지면 이러한 종류의 트레이스백이 매우 길어질 수 있습니다.
상황에 따라 `Plain` 또는 `Context` 모드의 간결함이 작업하기 더 쉬울 때도 있습니다.

## 디버깅: 트레이스백을 읽는 것만으로는 충분하지 않을 때

대화형 디버깅을 위한 표준 파이썬 도구는 `pdb`, 즉 파이썬 디버거입니다.
이 디버거를 사용하면 사용자가 코드를 한 줄씩 단계별로 실행하여 더 어려운 오류의 원인을 파악할 수 있습니다.
이것의 IPython 강화 버전은 `ipdb`, 즉 IPython 디버거입니다.

이 두 디버거를 시작하고 사용하는 방법에는 여러 가지가 있습니다. 여기서는 모두 다루지는 않겠습니다.
자세한 내용은 이 두 유틸리티의 온라인 설명서를 참조하십시오.

IPython에서 디버깅에 가장 편리한 인터페이스는 아마도 `%debug` 매직 명령어일 것입니다.
예외가 발생한 후 호출하면 예외 지점에서 대화형 디버깅 프롬프트가 자동으로 열립니다.
`ipdb` 프롬프트를 사용하면 스택의 현재 상태를 탐색하고 사용 가능한 변수를 탐색하며 파이썬 명령도 실행할 수 있습니다!

가장 최근 예외를 살펴본 다음 몇 가지 기본 작업을 수행해 보겠습니다. `a`와 `b`의 값을 인쇄한 다음 `quit`를 입력하여 디버깅 세션을 종료합니다.

```python
%debug
```

대화형 디버거는 이보다 훨씬 많은 기능을 제공합니다. 스택을 위아래로 이동하며 변수 값을 탐색할 수도 있습니다.

```python
%debug
```

이를 통해 오류의 원인뿐만 아니라 오류로 이어진 함수 호출도 신속하게 파악할 수 있습니다.

예외가 발생할 때마다 디버거가 자동으로 시작되도록 하려면 `%pdb` 매직 함수를 사용하여 이 자동 동작을 켤 수 있습니다.

```python
%xmode Plain
%pdb on
func2(1)
```

마지막으로, 대화형 모드에서 처음부터 실행하고 싶은 스크립트가 있는 경우 `%run -d` 명령으로 실행하고 `next` 명령을 사용하여 코드 줄을 대화형으로 단계별로 실행할 수 있습니다.

### 디버깅 명령어 부분 목록

여기에 표시된 것보다 대화형 디버깅에 사용할 수 있는 명령어가 훨씬 많습니다. 다음 표에는 더 일반적이고 유용한 명령어 중 일부에 대한 설명이 포함되어 있습니다:

| 명령어       |  설명                                                |
|---------------|-------------------------------------------------------------|
| `l(ist)`      | 파일의 현재 위치 표시                       |
| `h(elp)`      | 명령어 목록 표시 또는 특정 명령어에 대한 도움말 찾기 |
| `q(uit)`      | 디버거 및 프로그램 종료                           |
| `c(ontinue)`  | 디버거 종료, 프로그램 계속                  |
| `n(ext)`      | 프로그램의 다음 단계로 이동                          |
| `<enter>`     | 이전 명령어 반복                                 |
| `p(rint)`     | 변수 인쇄                                             |
| `s(tep)`      | 서브루틴으로 들어가기                                      |
| `r(eturn)`    | 서브루틴에서 나오기                                  |

자세한 내용은 디버거에서 `help` 명령어를 사용하거나 `ipdb`의 [온라인 설명서](https://github.com/gotcha/ipdb)를 참조하십시오.
