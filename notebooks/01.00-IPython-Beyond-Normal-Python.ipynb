# Jupyter: 일반 파이썬을 넘어서

파이썬 개발 환경에는 많은 옵션이 있으며, 저는 종종 제 작업에서 어떤 것을 사용하는지 질문을 받습니다.
제 대답은 때때로 사람들을 놀라게 합니다: 제가 선호하는 환경은 [IPython](http://ipython.org/)과 텍스트 편집기(제 경우에는 기분에 따라 Emacs 또는 VSCode)입니다.
Jupyter는 2001년 페르난도 페레즈가 향상된 파이썬 인터프리터로 만든 IPython 셸로 시작하여, 페레즈의 말에 따르면 "연구 컴퓨팅의 전체 수명 주기를 위한 도구"를 제공하는 것을 목표로 하는 프로젝트로 성장했습니다.
파이썬이 우리 데이터 과학 작업의 엔진이라면, Jupyter는 대화형 제어판이라고 생각할 수 있습니다.

파이썬에 대한 유용한 대화형 인터페이스일 뿐만 아니라 Jupyter는 언어에 여러 가지 유용한 구문 추가 기능도 제공합니다. 여기서는 이러한 추가 기능 중 가장 유용한 것을 다룰 것입니다.
아마도 Jupyter 프로젝트에서 제공하는 가장 친숙한 인터페이스는 Jupyter Notebook일 것입니다. 이는 데이터 과학 결과의 개발, 협업, 공유 및 게시에도 유용한 브라우저 기반 환경입니다.
노트북 형식의 유용성에 대한 예로, 현재 읽고 있는 페이지를 살펴보십시오. 이 책의 전체 원고는 Jupyter 노트북 세트로 구성되었습니다.

이 책의 이 부분에서는 데이터 과학 실습에 유용한 Jupyter 및 IPython 기능 중 일부를 단계별로 살펴보고, 특히 파이썬의 표준 기능을 넘어 제공하는 구문에 중점을 둘 것입니다.
다음으로, 데이터 과학 코드를 만들고 사용하는 일반적인 작업을 가속화할 수 있는 더 유용한 *매직 명령어* 중 일부에 대해 좀 더 자세히 살펴보겠습니다.
마지막으로, 데이터를 이해하고 결과를 공유하는 데 유용한 노트북의 몇 가지 기능을 살펴보겠습니다.
