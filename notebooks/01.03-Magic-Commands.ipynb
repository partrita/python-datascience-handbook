# IPython 매직 명령어

이전 장에서는 IPython을 사용하여 파이썬을 효율적이고 대화식으로 사용하고 탐색하는 방법을 보여주었습니다.
여기서는 IPython이 일반적인 파이썬 구문 위에 추가하는 몇 가지 향상된 기능에 대해 논의하기 시작하겠습니다.
이러한 기능은 IPython에서 *매직 명령어*로 알려져 있으며 `%` 문자로 시작합니다.
이러한 매직 명령어는 표준 데이터 분석에서 다양한 일반적인 문제를 간결하게 해결하도록 설계되었습니다.
매직 명령어에는 두 가지 종류가 있습니다. 단일 `%` 접두사로 표시되고 단일 입력 줄에서 작동하는 *라인 매직*과 이중 `%%` 접두사로 표시되고 여러 입력 줄에서 작동하는 *셀 매직*입니다.
여기서는 몇 가지 간단한 예를 시연하고 논의한 다음 나중에 몇 가지 유용한 매직 명령어에 대한 보다 집중적인 논의로 돌아올 것입니다.

## 외부 코드 실행: %run
더 광범위한 코드를 개발하기 시작하면 대화형 탐색을 위해 IPython을 사용하고 재사용하려는 코드를 저장하기 위해 텍스트 편집기를 사용하는 자신을 발견하게 될 것입니다.
이 코드를 새 창에서 실행하는 대신 IPython 세션 내에서 실행하는 것이 편리할 수 있습니다.
`%run` 매직 명령어로 이 작업을 수행할 수 있습니다.

예를 들어, 다음과 같은 내용으로 *myscript.py* 파일을 만들었다고 가정해 보겠습니다:

```python
# file: myscript.py

def square(x):
    """숫자를 제곱합니다"""
    return x ** 2

for N in range(1, 4):
    print(f"{N}의 제곱은 {square(N)}입니다")
```

IPython 세션에서 다음과 같이 실행할 수 있습니다:

```ipython
In [6]: %run myscript.py
1의 제곱은 1입니다
2의 제곱은 4입니다
3의 제곱은 9입니다
```

또한 이 스크립트를 실행한 후에는 스크립트 내에 정의된 모든 함수를 IPython 세션에서 사용할 수 있습니다:

```ipython
In [7]: square(5)
Out[7]: 25
```

코드가 실행되는 방식을 미세 조정하는 몇 가지 옵션이 있습니다. IPython 인터프리터에 **`%run?`**을 입력하여 일반적인 방법으로 설명서를 볼 수 있습니다.

## 코드 실행 시간 측정: %timeit
유용한 매직 함수의 또 다른 예는 `%timeit`이며, 이는 뒤따르는 단일 라인 파이썬 문의 실행 시간을 자동으로 결정합니다.
예를 들어, 리스트 컴프리헨션의 성능을 확인하고 싶을 수 있습니다:

```ipython
In [8]: %timeit L = [n ** 2 for n in range(1000)]
루프당 430 µs ± 3.21 µs (7회 실행, 각 1000 루프의 평균 ± 표준 편차)
```

`%timeit`의 장점은 짧은 명령의 경우 더 강력한 결과를 얻기 위해 자동으로 여러 번 실행한다는 것입니다.
여러 줄 문의 경우 두 번째 `%` 기호를 추가하면 여러 줄 입력을 처리할 수 있는 셀 매직으로 바뀝니다.
예를 들어, 다음은 `for` 루프를 사용한 동일한 구성입니다:

```ipython
In [9]: %%timeit
   ...: L = []
   ...: for n in range(1000):
   ...:     L.append(n ** 2)
   ...: 
루프당 484 µs ± 5.67 µs (7회 실행, 각 1000 루프의 평균 ± 표준 편차)
```

이 경우 리스트 컴프리헨션이 동등한 `for` 루프 구성보다 약 10% 빠르다는 것을 즉시 알 수 있습니다.
`%timeit` 및 코드 타이밍 및 프로파일링에 대한 다른 접근 방식은 [코드 프로파일링 및 타이밍](01.07-Timing-and-Profiling.ipynb)에서 살펴볼 것입니다.

## 매직 함수 도움말: ?, %magic 및 %lsmagic

일반적인 파이썬 함수와 마찬가지로 IPython 매직 함수에는 독스트링이 있으며 이 유용한
설명서는 표준적인 방식으로 액세스할 수 있습니다.
따라서 예를 들어 `%timeit` 매직 함수의 설명서를 읽으려면 다음과 같이 입력하면 됩니다:

```ipython
In [10]: %timeit?
```

다른 함수에 대한 설명서도 비슷하게 액세스할 수 있습니다.
몇 가지 예를 포함하여 사용 가능한 매직 함수에 대한 일반적인 설명을 보려면 다음과 같이 입력하면 됩니다:

```ipython
In [11]: %magic
```

사용 가능한 모든 매직 함수의 빠르고 간단한 목록을 보려면 다음과 같이 입력하십시오:

```ipython
In [12]: %lsmagic
```

마지막으로, 원한다면 자신만의 매직 함수를 정의하는 것이 매우 간단하다는 것을 언급하겠습니다.
