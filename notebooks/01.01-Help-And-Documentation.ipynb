# IPython 및 Jupyter 시작하기

데이터 과학을 위한 파이썬 코드를 작성할 때, 저는 일반적으로 세 가지 작업 모드를 오갑니다: 짧은 명령어 시퀀스를 시험해 보기 위해 IPython 셸을 사용하고, 더 긴 대화형 분석 및 다른 사람들과 콘텐츠를 공유하기 위해 Jupyter Notebook을 사용하며, 재사용 가능한 파이썬 패키지를 만들기 위해 Emacs 또는 VSCode와 같은 대화형 개발 환경(IDE)을 사용합니다.
이 장에서는 처음 두 가지 모드인 IPython 셸과 Jupyter Notebook에 중점을 둡니다.
소프트웨어 개발을 위한 IDE 사용은 데이터 과학자의 레퍼토리에서 중요한 세 번째 도구이지만, 여기서는 직접 다루지 않겠습니다.

## IPython 셸 실행하기

이 책의 대부분과 마찬가지로 이 부분의 텍스트는 수동적으로 흡수되도록 설계되지 않았습니다.
읽으면서 우리가 다루는 도구와 구문을 따라하고 실험해 보는 것이 좋습니다. 이렇게 함으로써 얻게 되는 근육 기억은 단순히 그것에 대해 읽는 것보다 훨씬 더 유용할 것입니다.
명령줄에 **`ipython`**을 입력하여 IPython 인터프리터를 실행하는 것으로 시작하십시오. 또는 Anaconda 또는 EPD와 같은 배포판을 설치한 경우 시스템에 특정한 실행기가 있을 수 있습니다([IPython 도움말 및 설명서](01.01-Help-And-Documentation.ipynb)에서 자세히 설명합니다).

이렇게 하면 다음과 같은 프롬프트가 표시됩니다:

```ipython
Python 3.9.2 (v3.9.2:1a79785e3e, Feb 19 2021, 09:06:10) 
Type 'copyright', 'credits' or 'license' for more information
IPython 7.21.0 -- An enhanced Interactive Python. Type '?' for help.

In [1]:
```
이제 따라갈 준비가 되었습니다.

## Jupyter Notebook 실행하기

Jupyter Notebook은 IPython 셸에 대한 브라우저 기반 그래픽 인터페이스이며, 풍부한 동적 디스플레이 기능을 기반으로 합니다.
파이썬/IPython 문을 실행하는 것 외에도 노트북을 사용하면 서식 있는 텍스트, 정적 및 동적 시각화, 수학 방정식, JavaScript 위젯 등을 포함할 수 있습니다.
또한 이러한 문서는 다른 사람들이 자신의 시스템에서 열고 코드를 실행할 수 있도록 저장할 수 있습니다.

웹 브라우저 창을 통해 Jupyter 노트북을 보고 편집하지만 코드를 실행하려면 실행 중인 파이썬 프로세스에 연결해야 합니다.
시스템 셸에서 다음 명령을 실행하여 이 프로세스("커널"이라고 함)를 시작할 수 있습니다:

```
$ jupyter lab
```

이 명령은 브라우저에 표시될 로컬 웹 서버를 시작합니다.
수행 중인 작업을 보여주는 로그를 즉시 출력합니다. 해당 로그는 다음과 같습니다:

```
$ jupyter lab
[ServerApp] Serving notebooks from local directory: /Users/jakevdp/PythonDataScienceHandbook
[ServerApp] Jupyter Server 1.4.1 is running at:
[ServerApp] http://localhost:8888/lab?token=dd852649
[ServerApp] Use Control-C to stop this server and shut down all kernels (twice to skip confirmation).
```

명령을 실행하면 기본 브라우저가 자동으로 열리고 나열된 로컬 URL로 이동해야 합니다.
정확한 주소는 시스템에 따라 다릅니다.
브라우저가 자동으로 열리지 않으면 창을 열고 이 주소(*http://localhost:8888/lab/* 이 예에서)를 수동으로 열 수 있습니다.

## IPython 도움말 및 설명서

이 장에서 다른 섹션을 읽지 않더라도 이 섹션은 읽으십시오. 여기서 논의된 도구는 IPython이 제 일상적인 워크플로에 가장 혁신적인 기여를 한 것이라고 생각합니다.

기술적으로 숙련된 사람이 친구, 가족 또는 동료의 컴퓨터 문제를 돕도록 요청받을 때, 대부분의 경우 답을 아는 것보다 알 수 없는 답을 신속하게 찾는 방법을 아는 것이 더 중요합니다.
데이터 과학에서도 마찬가지입니다. 온라인 설명서, 메일링 리스트 스레드 및 스택 오버플로 답변과 같은 검색 가능한 웹 리소스에는 이전에 검색해 본 적이 있는 주제에 대해서도 (특히?) 풍부한 정보가 포함되어 있습니다.
효과적인 데이터 과학 실무자가 되는 것은 모든 가능한 상황에 사용해야 하는 도구나 명령을 암기하는 것보다 웹 검색 엔진이나 다른 수단을 통해 모르는 정보를 효과적으로 찾는 방법을 배우는 것입니다.

IPython/Jupyter의 가장 유용한 기능 중 하나는 사용자와 효과적으로 작업을 수행하는 데 도움이 되는 설명서 및 검색 유형 간의 간격을 줄이는 것입니다.
웹 검색은 여전히 복잡한 질문에 답하는 데 중요한 역할을 하지만, IPython만으로도 놀라운 양의 정보를 찾을 수 있습니다.
IPython이 몇 번의 키 입력으로 답을 찾는 데 도움이 될 수 있는 질문의 예는 다음과 같습니다:

- 이 함수를 어떻게 호출합니까? 어떤 인수와 옵션이 있습니까?
- 이 파이썬 객체의 소스 코드는 어떻게 생겼습니까?
- 가져온 이 패키지에는 무엇이 있습니까? 
- 이 객체에는 어떤 속성이나 메서드가 있습니까?

여기서는 IPython 셸과 Jupyter Notebook에서 이 정보에 신속하게 액세스하기 위해 제공되는 도구, 즉 설명서를 탐색하기 위한 `?` 문자, 소스 코드를 탐색하기 위한 `??` 문자, 자동 완성을 위한 Tab 키에 대해 설명합니다.

### ?로 문서 접근하기

파이썬 언어와 데이터 과학 생태계는 사용자를 염두에 두고 만들어졌으며, 그 중 큰 부분은 문서에 대한 접근입니다.
모든 파이썬 객체에는 *독스트링*이라고 하는 문자열에 대한 참조가 포함되어 있으며, 대부분의 경우 객체와 사용 방법에 대한 간결한 요약이 포함됩니다.
파이썬에는 이 정보에 접근하여 결과를 인쇄할 수 있는 내장 `help` 함수가 있습니다.
예를 들어, 내장 `len` 함수의 설명서를 보려면 다음을 수행할 수 있습니다:

```ipython
In [1]: help(len)
Help on built-in function len in module builtins:

len(obj, /)
    Return the number of items in a container.
```

인터프리터에 따라 이 정보는 인라인 텍스트로 표시되거나 별도의 팝업 창에 표시될 수 있습니다.

객체에 대한 도움말을 찾는 것이 매우 일반적이고 유용하기 때문에 IPython과 Jupyter는 이 설명서 및 기타 관련 정보에 액세스하기 위한 약식으로 `?` 문자를 도입합니다:

```ipython
In [2]: len?
Signature: len(obj, /)
Docstring: Return the number of items in a container.
Type:      builtin_function_or_method
```

이 표기법은 객체 메서드를 포함하여 거의 모든 것에 적용됩니다:

```ipython
In [3]: L = [1, 2, 3]
In [4]: L.insert?
Signature: L.insert(index, object, /)
Docstring: Insert object before index.
Type:      builtin_function_or_method
```

또는 해당 유형의 설명서와 함께 객체 자체도 가능합니다:

```ipython
In [5]: L?
Type:        list
String form: [1, 2, 3]
Length:      3
Docstring:  
Built-in mutable sequence.

If no argument is given, the constructor creates a new empty list.
The argument must be an iterable if specified.
```

중요한 것은, 이것은 여러분이 직접 만든 함수나 다른 객체에도 적용된다는 것입니다!
여기서는 독스트링이 있는 작은 함수를 정의하겠습니다:

```ipython
In [6]: def square(a):
  ....:     """a의 제곱을 반환합니다."""
  ....:     return a ** 2
  ....:
```

함수에 대한 독스트링을 만들기 위해 첫 번째 줄에 문자열 리터럴을 배치했습니다.
독스트링은 일반적으로 여러 줄이므로 관례적으로 여러 줄 문자열에 대한 파이썬의 삼중 따옴표 표기법을 사용했습니다.

이제 `?`를 사용하여 이 독스트링을 찾습니다:

```ipython
In [7]: square?
Signature: square(a)
Docstring: a의 제곱을 반환합니다.
File:      <ipython-input-6>
Type:      function
```

독스트링을 통해 설명서에 빠르게 액세스할 수 있다는 점이 바로 여러분이 작성하는 코드에 항상 이러한 인라인 설명서를 추가하는 습관을 들여야 하는 이유 중 하나입니다!

### ??로 소스 코드 접근하기

파이썬 언어는 매우 읽기 쉽기 때문에 궁금한 객체의 소스 코드를 읽으면 일반적으로 또 다른 수준의 통찰력을 얻을 수 있습니다.
IPython과 Jupyter는 이중 물음표(`??`)를 사용하여 소스 코드에 대한 바로 가기를 제공합니다:

```ipython
In [8]: square??
Signature: square(a)
Source:   
def square(a):
    """a의 제곱을 반환합니다."""
    return a ** 2
File:      <ipython-input-6>
Type:      function
```

이와 같은 간단한 함수의 경우 이중 물음표를 사용하면 내부 세부 정보에 대한 빠른 통찰력을 얻을 수 있습니다.

이것을 많이 사용해 보면 `??` 접미사가 소스 코드를 표시하지 않는 경우가 있다는 것을 알게 될 것입니다. 이는 일반적으로 해당 객체가 파이썬이 아닌 C 또는 다른 컴파일된 확장 언어로 구현되었기 때문입니다.
이 경우 `??` 접미사는 `?` 접미사와 동일한 출력을 제공합니다.
이것은 이전에 사용했던 `len` 함수를 포함하여 파이썬의 많은 내장 객체 및 유형에서 특히 많이 볼 수 있습니다:

```ipython
In [9]: len??
Signature: len(obj, /)
Docstring: Return the number of items in a container.
Type:      builtin_function_or_method
```

`?` 및/또는 `??`를 사용하는 것은 파이썬 함수 또는 모듈이 수행하는 작업에 대한 정보를 찾는 강력하고 빠른 방법입니다.

### 탭 완성을 사용하여 모듈 탐색하기

또 다른 유용한 인터페이스는 객체, 모듈 및 네임스페이스의 내용을 자동 완성하고 탐색하기 위해 Tab 키를 사용하는 것입니다.
다음 예에서는 Tab 키를 눌러야 할 때 `<TAB>`을 사용합니다.

#### 객체 내용의 탭 완성

모든 파이썬 객체에는 다양한 속성과 메서드가 연결되어 있습니다.
앞서 언급한 `help` 함수와 마찬가지로 파이썬에는 이러한 목록을 반환하는 내장 `dir` 함수가 있지만 탭 완성 인터페이스는 실제로 사용하기가 훨씬 쉽습니다.
객체의 사용 가능한 모든 속성 목록을 보려면 객체 이름 뒤에 마침표(".") 문자와 Tab 키를 입력하면 됩니다:

```ipython
In [10]: L.<TAB>
            append() count    insert   reverse 
            clear    extend   pop      sort    
            copy     index    remove           
```

목록을 좁히려면 이름의 첫 번째 문자 또는 여러 문자를 입력하면 Tab 키가 일치하는 속성과 메서드를 찾습니다:

```ipython
In [10]: L.c<TAB>
             clear() count()
             copy()         

In [10]: L.co<TAB>
              copy()  count()
```

옵션이 하나만 있는 경우 Tab 키를 누르면 해당 줄이 자동으로 완성됩니다.
예를 들어, 다음은 즉시 `L.count`로 바뀝니다:

```ipython
In [10]: L.cou<TAB>

```

파이썬에는 공용/외부 속성과 개인/내부 속성 간에 엄격하게 적용되는 구분이 없지만, 관례적으로 후자를 나타내는 데 앞에 밑줄을 사용합니다.
명확성을 위해 이러한 개인 메서드와 특수 메서드는 기본적으로 목록에서 생략되지만 밑줄을 명시적으로 입력하여 나열할 수 있습니다:

```ipython
In [10]: L._<TAB>
           __add__             __delattr__     __eq__      
           __class__           __delitem__     __format__()
           __class_getitem__() __dir__()       __ge__            >
           __contains__        __doc__         __getattribute__     
```

간결성을 위해 출력의 처음 몇 개 열만 표시했습니다.
이들 대부분은 파이썬의 특수 이중 밑줄 메서드(종종 "던더" 메서드라고 함)입니다.

#### 가져올 때 탭 완성

탭 완성은 패키지에서 객체를 가져올 때도 유용합니다.
여기서는 `itertools` 패키지에서 `co`로 시작하는 모든 가능한 가져오기를 찾는 데 사용합니다:

```ipython
In [10]: from itertools import co<TAB>
         combinations()                  compress()
         combinations_with_replacement() count()
```

마찬가지로 탭 완성을 사용하여 시스템에서 사용할 수 있는 가져오기를 확인할 수 있습니다(파이썬 세션에 표시되는 타사 스크립트 및 모듈에 따라 변경됨):

```ipython
In [10]: import <TAB>
            abc                 anyio                          
            activate_this       appdirs                        
            aifc                appnope        >
            antigravity         argon2                         

In [10]: import h<TAB>
            hashlib html   
            heapq   http   
            hmac        
```

#### 탭 완성을 넘어: 와일드카드 일치

탭 완성은 찾고 있는 객체 또는 속성 이름의 처음 몇 글자를 알고 있는 경우 유용하지만, 이름 중간이나 끝에 있는 문자를 일치시키려는 경우에는 거의 도움이 되지 않습니다.
이러한 사용 사례를 위해 IPython과 Jupyter는 `*` 문자를 사용하여 이름에 대한 와일드카드 일치 수단을 제공합니다.

예를 들어, 이를 사용하여 이름이 `Warning`으로 끝나는 네임스페이스의 모든 객체를 나열할 수 있습니다:

```ipython
In [10]: *Warning?
BytesWarning                  RuntimeWarning
DeprecationWarning            SyntaxWarning
FutureWarning                 UnicodeWarning
ImportWarning                 UserWarning
PendingDeprecationWarning     Warning
ResourceWarning
```

`*` 문자는 빈 문자열을 포함하여 모든 문자열과 일치합니다.

마찬가지로, 이름 어딘가에 `find`라는 단어가 포함된 문자열 메서드를 찾고 있다고 가정해 보겠습니다.
다음과 같이 검색할 수 있습니다:

```ipython
In [11]: str.*find*?
str.find
str.rfind
```

이러한 유형의 유연한 와일드카드 검색은 새 패키지를 익히거나 익숙한 패키지를 다시 익힐 때 특정 명령을 찾는 데 유용할 수 있다고 생각합니다.
