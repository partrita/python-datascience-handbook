# 입력 및 출력 기록

이전에 IPython 셸에서 위쪽 및 아래쪽 화살표 키 또는 Ctrl-p/Ctrl-n 단축키를 사용하여 이전 명령에 액세스할 수 있음을 확인했습니다.
또한 셸과 노트북 모두에서 IPython은 이전 명령의 출력과 명령 자체의 문자열 버전을 가져오는 여러 가지 방법을 제공합니다.
여기서 살펴보겠습니다.

## IPython의 In 및 Out 객체

이제 IPython에서 사용하는 `In [1]:`/`Out[1]:` 스타일의 프롬프트에 익숙해졌을 것이라고 생각합니다.
그러나 이것들은 단지 예쁜 장식이 아닙니다. 현재 세션에서 이전 입력 및 출력에 액세스하는 방법에 대한 단서를 제공합니다.
다음과 같은 세션을 시작한다고 가정해 보겠습니다:

```ipython
In [1]: import math

In [2]: math.sin(2)
Out[2]: 0.9092974268256817

In [3]: math.cos(2)
Out[3]: -0.4161468365471424
```

내장 `math` 패키지를 가져온 다음 숫자 2의 사인과 코사인을 계산했습니다.
이러한 입력과 출력은 셸에 `In`/`Out` 레이블과 함께 표시되지만 IPython은 실제로 이 기록을 반영하도록 자동으로 업데이트되는 `In` 및 `Out`이라는 일부 파이썬 변수를 만듭니다:

```ipython
In [4]: In
Out[4]: ['', 'import math', 'math.sin(2)', 'math.cos(2)', 'In']

In [5]: Out
Out[5]:
{2: 0.9092974268256817,
 3: -0.4161468365471424,
 4: ['', 'import math', 'math.sin(2)', 'math.cos(2)', 'In', 'Out']}
```

`In` 객체는 명령을 순서대로 추적하는 목록입니다(목록의 첫 번째 항목은 `In [1]`이 첫 번째 명령을 참조할 수 있도록 하는 자리 표시자입니다):

```ipython
In [6]: print(In[1])
import math
```

`Out` 객체는 목록이 아니라 입력 번호를 해당 출력(있는 경우)에 매핑하는 사전입니다:

```ipython
In [7]: print(Out[2])
0.9092974268256817
```

모든 작업에 출력이 있는 것은 아닙니다. 예를 들어 `import` 문과 `print` 문은 출력에 영향을 미치지 않습니다.
후자는 놀랍게 보일 수 있지만 `print`가 `None`을 반환하는 함수라는 점을 고려하면 이해가 됩니다. 간결성을 위해 `None`을 반환하는 모든 명령은 `Out`에 추가되지 않습니다.

이것이 유용할 수 있는 경우는 과거 결과와 상호 작용하려는 경우입니다.
예를 들어 이전에 계산된 결과를 사용하여 `sin(2) ** 2`와 `cos(2) ** 2`의 합계를 확인해 보겠습니다:

```ipython
In [8]: Out[2] ** 2 + Out[3] ** 2
Out[8]: 1.0
```

결과는 잘 알려진 삼각 항등식에서 예상한 대로 `1.0`입니다.
이 경우 이러한 이전 결과를 사용하는 것이 반드시 필요한 것은 아니지만 매우 비용이 많이 드는 계산을 실행하고 결과를 변수에 할당하는 것을 잊은 경우 매우 유용할 수 있습니다.

## 밑줄 단축키 및 이전 출력

표준 파이썬 셸에는 이전 출력에 액세스하기 위한 간단한 단축키가 하나만 포함되어 있습니다. 변수 `_`(즉, 단일 밑줄)는 이전 출력으로 계속 업데이트됩니다. 이것은 IPython에서도 작동합니다:

```ipython
In [9]: print(_)
1.0
```

그러나 IPython은 이것을 좀 더 발전시켰습니다. 이중 밑줄을 사용하여 마지막에서 두 번째 출력에 액세스하고 삼중 밑줄을 사용하여 마지막에서 세 번째 출력에 액세스할 수 있습니다(출력이 없는 명령은 건너뜁니다):

```ipython
In [10]: print(__)
-0.4161468365471424

In [11]: print(___)
0.9092974268256817
```

IPython은 거기서 멈춥니다. 밑줄이 세 개 이상이면 세기가 약간 어려워지고 그 시점에서는 줄 번호로 출력을 참조하는 것이 더 쉽습니다.

그러나 언급해야 할 단축키가 하나 더 있습니다. `Out[X]`의 약어는 `_X`(즉, 단일 밑줄 뒤에 줄 번호)입니다:

```ipython
In [12]: Out[2]
Out[12]: 0.9092974268256817

In [13]: _2
Out[13]: 0.9092974268256817
```

## 출력 억제
때로는 문의 출력을 억제하고 싶을 수 있습니다([Matplotlib 소개](04.00-Introduction-To-Matplotlib.ipynb)에서 살펴볼 플로팅 명령에서 가장 일반적일 수 있습니다).
또는 실행 중인 명령이 출력 기록에 저장하고 싶지 않은 결과를 생성할 수도 있습니다. 아마도 다른 참조가 제거될 때 할당 해제될 수 있도록 하기 위해서일 것입니다.
명령의 출력을 억제하는 가장 쉬운 방법은 줄 끝에 세미콜론을 추가하는 것입니다:

```ipython
In [14]: math.sin(2) + math.cos(2);
```

결과는 자동으로 계산되며 출력은 화면에 표시되거나 `Out` 사전에 저장되지 않습니다:

```ipython
In [15]: 14 in Out
Out[15]: False
```

## 관련된 매직 명령어
한 번에 이전 입력의 일괄 처리에 액세스하려면 `%history` 매직 명령어가 매우 유용합니다.
다음은 처음 네 개의 입력을 인쇄하는 방법입니다:

```ipython
In [16]: %history -n 1-3
   1: import math
   2: math.sin(2)
   3: math.cos(2)
```

평소와 같이 `%history?`를 입력하여 더 많은 정보와 사용 가능한 옵션에 대한 설명을 볼 수 있습니다(`?` 기능에 대한 자세한 내용은 [IPython의 도움말 및 설명서](01.01-Help-And-Documentation.ipynb) 참조).
다른 유용한 매직 명령어로는 명령어 기록의 일부를 다시 실행하는 `%rerun`과 명령어 기록의 일부 집합을 파일에 저장하는 `%save`가 있습니다.
