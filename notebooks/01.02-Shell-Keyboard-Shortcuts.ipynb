# IPython 셸의 키보드 단축키

컴퓨터에서 시간을 보내는 사람이라면 누구나 워크플로에서 키보드 단축키를 사용해 본 적이 있을 것입니다.
가장 익숙한 것은 아마도 다양한 프로그램과 시스템에서 복사 및 붙여넣기에 사용되는 Cmd-c 및 Cmd-v(또는 Ctrl-c 및 Ctrl-v)일 것입니다.
파워 유저는 훨씬 더 나아가기도 합니다. Emacs, Vim 등과 같은 인기 있는 텍스트 편집기는 복잡한 키 입력 조합을 통해 사용자에게 놀라운 범위의 작업을 제공합니다.

IPython 셸은 그렇게까지 하지는 않지만 명령을 입력하는 동안 빠른 탐색을 위한 여러 키보드 단축키를 제공합니다.
이러한 단축키 중 일부는 브라우저 기반 노트북에서도 작동하지만 이 섹션은 주로 IPython 셸의 단축키에 관한 것입니다.

이러한 단축키에 익숙해지면 "홈" 키보드 위치에서 손을 떼지 않고도 특정 명령을 빠르게 수행하는 데 매우 유용할 수 있습니다.
Emacs 사용자이거나 Linux 스타일 셸 경험이 있다면 다음 내용이 매우 익숙할 것입니다.
이러한 단축키를 *탐색 단축키*, *텍스트 입력 단축키*, *명령 기록 단축키* 및 *기타 단축키*의 몇 가지 범주로 그룹화합니다.

## 탐색 단축키

줄에서 앞뒤로 이동하기 위해 왼쪽 및 오른쪽 화살표 키를 사용하는 것은 매우 분명하지만 "홈" 키보드 위치에서 손을 떼지 않아도 되는 다른 옵션이 있습니다:

| 키 입력                         | 작업                                     |
|---------------------------------|--------------------------------------------|
| Ctrl-a                          | 줄 시작으로 커서 이동           |
| Ctrl-e                          | 줄 끝으로 커서 이동             |
| Ctrl-b 또는 왼쪽 화살표 키    | 커서를 한 문자 뒤로 이동             |
| Ctrl-f 또는 오른쪽 화살표 키   | 커서를 한 문자 앞으로 이동          |

## 텍스트 입력 단축키

이전 문자를 삭제하기 위해 백스페이스 키를 사용하는 것은 누구나 익숙하지만 키에 도달하려면 약간의 손가락 체조가 필요한 경우가 많으며 한 번에 한 문자만 삭제합니다.
IPython에는 입력 중인 텍스트의 일부를 제거하기 위한 몇 가지 단축키가 있습니다. 이 중 가장 즉시 유용한 것은 텍스트 줄 전체를 삭제하는 명령입니다.
이전 문자를 삭제하기 위해 백스페이스 키에 도달하는 대신 Ctrl-b와 Ctrl-d의 조합을 사용하는 자신을 발견하면 이러한 단축키가 제2의 천성이 되었다는 것을 알게 될 것입니다!

| 키 입력                   | 작업                                           |
|-----------------------------|--------------------------------------------------|
| 백스페이스 키               | 줄의 이전 문자 삭제                |
| Ctrl-d                      | 줄의 다음 문자 삭제                    |
| Ctrl-k                      | 커서에서 줄 끝까지 텍스트 잘라내기              |
| Ctrl-u                      | 줄 시작부터 커서까지 텍스트 잘라내기        |
| Ctrl-y                      | 이전에 잘라낸 텍스트 붙여넣기(즉, 붙여넣기)  |
| Ctrl-t                      | 이전 두 문자 바꾸기(즉, 전환) |

## 명령어 히스토리 단축키

여기서 논의된 단축키 중 아마도 가장 영향력 있는 것은 IPython이 명령어 히스토리 탐색을 위해 제공하는 단축키일 것입니다.
이 명령어 히스토리는 현재 IPython 세션을 넘어섭니다. 전체 명령어 히스토리는 IPython 프로필 디렉토리의 SQLite 데이터베이스에 저장됩니다.
이전 명령어에 액세스하는 가장 간단한 방법은 위쪽 및 아래쪽 화살표 키를 사용하여 히스토리를 단계별로 살펴보는 것이지만 다른 옵션도 있습니다:

| 키 입력                           | 작업                                     |
|-----------------------------------|--------------------------------------------|
| Ctrl-p (또는 위쪽 화살표 키)      | 히스토리에서 이전 명령어 액세스         |
| Ctrl-n (또는 아래쪽 화살표 키)    | 히스토리에서 다음 명령어 액세스             |
| Ctrl-r                            | 명령어 히스토리를 통한 역방향 검색     |

역검색 옵션은 특히 유용할 수 있습니다.
앞서 `square`라는 함수를 정의했음을 기억하십시오.
새로운 IPython 셸에서 파이썬 기록을 역검색하여 이 정의를 다시 찾아보겠습니다.
IPython 터미널에서 Ctrl-r을 누르면 다음 프롬프트가 표시됩니다:

```ipython
In [1]:
(reverse-i-search)`': 
```

이 프롬프트에서 문자를 입력하기 시작하면 IPython은 해당 문자와 일치하는 가장 최근 명령(있는 경우)을 자동으로 채웁니다:

```ipython
In [1]: 
(reverse-i-search)`sqa': square??
```

언제든지 검색을 구체화하기 위해 더 많은 문자를 추가하거나 Ctrl-r을 다시 눌러 쿼리와 일치하는 다른 명령을 추가로 검색할 수 있습니다. 이전에 따라했다면 Ctrl-r을 두 번 더 누르면 다음과 같이 표시됩니다:

```ipython
In [1]: 
(reverse-i-search)`sqa': def square(a):
    """a의 제곱을 반환합니다"""
    return a ** 2
```

찾고 있는 명령을 찾았으면 Return을 누르면 검색이 종료됩니다.
그런 다음 검색된 명령을 사용하고 세션을 계속할 수 있습니다:

```ipython
In [1]: def square(a):
    """a의 제곱을 반환합니다"""
    return a ** 2

In [2]: square(2)
Out[2]: 4
```

Ctrl-p/Ctrl-n 또는 위/아래 화살표 키를 사용하여 비슷한 방식으로 기록을 검색할 수 있지만 줄 시작 부분의 문자만 일치시킵니다.
즉, **`def`**를 입력한 다음 Ctrl-p를 누르면 기록에서 `def` 문자로 시작하는 가장 최근 명령(있는 경우)을 찾습니다.

## 기타 단축키

마지막으로, 이전 범주에 속하지 않지만 알아두면 유용한 몇 가지 기타 단축키가 있습니다:

| 키 입력                   | 작업                                     |
|-----------------------------|--------------------------------------------|
| Ctrl-l                      | 터미널 화면 지우기                      |
| Ctrl-c                      | 현재 파이썬 명령 중단           |
| Ctrl-d                      | IPython 세션 종료                       |

특히 Ctrl-c 단축키는 실수로 매우 오래 실행되는 작업을 시작했을 때 유용할 수 있습니다.

여기서 논의된 단축키 중 일부는 처음에는 다소 모호해 보일 수 있지만 연습을 통해 빠르게 자동화됩니다.
그 근육 기억을 개발하면 다른 상황에서도 사용할 수 있기를 바라는 자신을 발견하게 될 것이라고 생각합니다.
