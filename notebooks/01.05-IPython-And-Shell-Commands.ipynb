# IPython 및 셸 명령어

표준 파이썬 인터프리터로 대화형으로 작업할 때 불편한 점 중 하나는 파이썬 도구와 시스템 명령줄 도구에 액세스하기 위해 여러 창 사이를 전환해야 한다는 것입니다.
IPython은 이러한 격차를 해소하고 IPython 터미널 내에서 직접 셸 명령을 실행하기 위한 구문을 제공합니다.
마법은 느낌표로 일어납니다. 줄에서 `!` 뒤에 나타나는 모든 것은 파이썬 커널이 아닌 시스템 명령줄에 의해 실행됩니다.

다음 논의는 Linux 또는 macOS와 같은 Unix 계열 시스템을 사용하고 있다고 가정합니다.
다음 예제 중 일부는 기본적으로 다른 유형의 셸을 사용하는 Windows에서 실패하지만, *Linux용 Windows 하위 시스템*을 사용하는 경우 여기 예제가 올바르게 실행되어야 합니다.
셸 명령에 익숙하지 않은 경우 항상 훌륭한 Software Carpentry Foundation에서 제공하는 [Unix 셸 자습서](http://swcarpentry.github.io/shell-novice/)를 검토하는 것이 좋습니다.

## 셸에 대한 간략한 소개

셸/터미널/명령줄 사용에 대한 전체 소개는 이 장의 범위를 훨씬 벗어나지만, 처음 접하는 사람들을 위해 여기서 간략한 소개를 제공하겠습니다.
셸은 컴퓨터와 텍스트로 상호 작용하는 방법입니다.
Microsoft와 Apple이 현재 보편화된 그래픽 운영 체제의 첫 번째 버전을 출시한 1980년대 중반 이후 대부분의 컴퓨터 사용자는 익숙한 메뉴 선택 및 끌어서 놓기 이동을 통해 운영 체제와 상호 작용해 왔습니다.
그러나 운영 체제는 이러한 그래픽 사용자 인터페이스보다 훨씬 오래 전에 존재했으며 주로 텍스트 입력 시퀀스를 통해 제어되었습니다. 프롬프트에서 사용자는 명령을 입력하고 컴퓨터는 사용자가 지시한 작업을 수행했습니다.
이러한 초기 프롬프트 시스템은 대부분의 데이터 과학자가 오늘날에도 여전히 사용하는 셸과 터미널의 전신이었습니다.

셸에 익숙하지 않은 사람은 아이콘과 메뉴를 클릭하기만 하면 동일한 결과를 많이 얻을 수 있는데 왜 굳이 이렇게 해야 하는지 물을 수 있습니다.
셸 사용자는 다른 질문으로 대답할 수 있습니다. 입력하여 훨씬 더 쉽게 작업을 수행할 수 있는데 왜 아이콘과 메뉴 항목을 찾아야 합니까?
일반적인 기술 선호도 교착 상태처럼 들릴 수 있지만 기본 작업을 넘어설 때 셸이 고급 작업에 훨씬 더 많은 제어를 제공한다는 것이 금방 분명해집니다. 물론 학습 곡선이 위협적일 수 있다는 점은 인정합니다.

예를 들어, 다음은 사용자가 시스템에서 디렉터리와 파일을 탐색, 생성 및 수정하는 Linux/macOS 셸 세션의 샘플입니다(`osx:~ $`는 프롬프트이고 `$` 뒤의 모든 것은 입력된 명령입니다. `#` 앞에 오는 텍스트는 실제로 입력하는 것이 아니라 설명일 뿐입니다):

```bash
osx:~ $ echo "hello world"             # echo는 파이썬의 print 함수와 같습니다
hello world

osx:~ $ pwd                            # pwd = 현재 작업 디렉터리 인쇄
/home/jake                             # 이것이 우리가 있는 "경로"입니다

osx:~ $ ls                             # ls = 현재 작업 디렉터리 내용 나열
notebooks  projects 

osx:~ $ cd projects/                   # cd = 디렉터리 변경

osx:projects $ pwd
/home/jake/projects

osx:projects $ ls
datasci_book   mpld3   myproject.txt

osx:projects $ mkdir myproject          # mkdir = 새 디렉터리 만들기

osx:projects $ cd myproject/

osx:myproject $ mv ../myproject.txt ./  # mv = 파일 이동. 여기서 우리는 파일을 이동하고 있습니다
                                        # myproject.txt 파일을 한 디렉터리 위로(../) 현재 디렉터리(./)로 이동합니다.
osx:myproject $ ls
myproject.txt
```

이 모든 것은 아이콘과 메뉴를 클릭하는 대신 명령을 입력하여 익숙한 작업(디렉터리 구조 탐색, 디렉터리 만들기, 파일 이동 등)을 수행하는 간결한 방법일 뿐입니다.
몇 가지 명령(`pwd`, `ls`, `cd`, `mkdir` 및 `cp`)만으로도 가장 일반적인 파일 작업을 많이 수행할 수 있지만 이러한 기본 사항을 넘어서면 셸 접근 방식이 정말 강력해집니다.

## IPython의 셸 명령어

모든 표준 셸 명령어는 `!` 문자를 앞에 붙여 IPython에서 직접 사용할 수 있습니다.
예를 들어 `ls`, `pwd` 및 `echo` 명령어는 다음과 같이 실행할 수 있습니다:

```ipython
In [1]: !ls
myproject.txt

In [2]: !pwd
/home/jake/projects/myproject

In [3]: !echo "셸에서 인쇄 중"
셸에서 인쇄 중
```

## 셸과 값 주고받기

셸 명령어는 IPython에서 호출할 수 있을 뿐만 아니라 IPython 네임스페이스와 상호 작용하도록 만들 수도 있습니다.
예를 들어 할당 연산자 `=`를 사용하여 모든 셸 명령어의 출력을 파이썬 목록에 저장할 수 있습니다:

```ipython
In [4]: contents = !ls

In [5]: print(contents)
['myproject.txt']

In [6]: directory = !pwd

In [7]: print(directory)
['/Users/jakevdp/notebooks/tmp/myproject']
```

이러한 결과는 목록으로 반환되는 것이 아니라 IPython에 정의된 특수 셸 반환 유형으로 반환됩니다:

```ipython
In [8]: type(directory)
IPython.utils.text.SList
```

이것은 파이썬 목록과 매우 유사하게 보이고 작동하지만 다음과 같은 추가 기능이 있습니다.
`grep` 및 `fields` 메서드와 `s`, `n` 및 `p` 속성을 사용하면 편리한 방식으로 결과를 검색, 필터링 및 표시할 수 있습니다.
이에 대한 자세한 내용은 IPython의 기본 제공 도움말 기능을 사용할 수 있습니다.

다른 방향, 즉 파이썬 변수를 셸로 전달하는 통신은 `{varname}` 구문을 사용하여 가능합니다:

```ipython
In [9]: message = "파이썬에서 온 안녕하세요"

In [10]: !echo {message}
파이썬에서 온 안녕하세요
```

중괄호에는 변수 이름이 포함되며 셸 명령에서 변수의 내용으로 바뀝니다.

## 셸 관련 매직 명령어

IPython의 셸 명령어를 잠시 사용해 보면 파일 시스템을 탐색하는 데 `!cd`를 사용할 수 없다는 것을 알 수 있습니다:

```ipython
In [11]: !pwd
/home/jake/projects/myproject

In [12]: !cd ..

In [13]: !pwd
/home/jake/projects/myproject
```

그 이유는 노트북의 셸 명령어가 명령어 간 상태를 유지하지 않는 임시 하위 셸에서 실행되기 때문입니다.
작업 디렉터리를 좀 더 영구적으로 변경하려면 `%cd` 매직 명령어를 사용할 수 있습니다:

```ipython
In [14]: %cd ..
/home/jake/projects
```

사실, 기본적으로 `%` 기호 없이도 사용할 수 있습니다:

```ipython
In [15]: cd myproject
/home/jake/projects/myproject
```

이것은 *자동 매직* 함수로 알려져 있으며 명시적인 `%` 없이 이러한 명령어를 실행하는 기능은 `%automagic` 매직 함수로 전환할 수 있습니다.

`%cd` 외에도 사용 가능한 다른 셸과 유사한 매직 함수로는 `%cat`, `%cp`, `%env`, `%ls`, `%man`, `%mkdir`, `%more`, `%mv`, `%pwd`, `%rm` 및 `%rmdir`이 있으며, `automagic`이 켜져 있으면 `%` 기호 없이 사용할 수 있습니다.
이렇게 하면 IPython 프롬프트를 일반 셸처럼 거의 처리할 수 있습니다:

```ipython
In [16]: mkdir tmp

In [17]: ls
myproject.txt  tmp/

In [18]: cp myproject.txt tmp/

In [19]: ls tmp
myproject.txt

In [20]: rm -r tmp
```

파이썬 세션과 동일한 터미널 창 내에서 셸에 액세스하면 컨텍스트 전환이 줄어들어 워크플로에서 파이썬과 셸을 보다 자연스럽게 결합할 수 있습니다.
