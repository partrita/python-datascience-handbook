# 서문

## 데이터 과학이란 무엇인가?

이 책은 파이썬으로 데이터 과학을 하는 것에 관한 책이며, 이는 즉시 다음과 같은 질문을 제기합니다: *데이터 과학*이란 무엇인가?
이 용어가 얼마나 보편화되었는지를 고려할 때 놀랍게도 정의하기 어려운 용어입니다.
비판적인 목소리들은 다양하게 이를 불필요한 레이블(결국 데이터를 포함하지 않는 과학이 어디 있겠는가?) 또는 이력서를 돋보이게 하고 과도하게 열성적인 기술 채용 담당자의 눈길을 사로잡기 위해 존재하는 단순한 유행어라고 일축했습니다.

제 생각에 이러한 비판은 중요한 것을 놓치고 있습니다.
데이터 과학은 과대 포장된 겉모습에도 불구하고, 산업 및 학계의 많은 응용 분야에서 점점 더 중요해지고 있는 학제 간 기술 집합에 대해 우리가 가진 가장 좋은 레이블일 것입니다.
이 *학제 간* 부분은 핵심입니다: 제 생각에 데이터 과학에 대한 가장 좋은 기존 정의는 2010년 9월 자신의 블로그에 처음 게시된 드류 콘웨이의 데이터 과학 벤 다이어그램으로 설명됩니다(다음 그림 참조).

![데이터 과학 벤 다이어그램](images/Data_Science_VD.png)

<small>(출처: [드류 콘웨이](http://drewconway.com/zia/2013/3/26/the-data-science-venn-diagram), 허가를 받아 사용)</small>

교차 영역 레이블 중 일부는 다소 농담조이지만, 이 다이어그램은 사람들이 "데이터 과학"이라고 말할 때 의미하는 바의 본질을 포착합니다: 그것은 근본적으로 학제 간 주제입니다.
데이터 과학은 세 가지 뚜렷하고 중복되는 영역으로 구성됩니다: 점점 더 커지고 있는 데이터 세트를 모델링하고 요약하는 방법을 아는 *통계학자*의 기술; 이 데이터를 효율적으로 저장, 처리 및 시각화하기 위한 알고리즘을 설계하고 사용할 수 있는 *컴퓨터 과학자*의 기술; 그리고 올바른 질문을 공식화하고 그 답변을 맥락에 맞게 배치하는 데 모두 필요한 *도메인 전문 지식*—우리가 어떤 주제에 대한 "고전적인" 훈련이라고 생각할 수 있는 것.

이를 염두에 두고, 데이터 과학을 배워야 할 새로운 지식 영역이 아니라 현재 전문 분야 내에서 적용할 수 있는 새로운 기술 집합으로 생각하도록 권장합니다.
선거 결과를 보고하든, 주식 수익률을 예측하든, 온라인 광고 클릭을 최적화하든, 현미경 사진에서 미생물을 식별하든, 새로운 종류의 천체를 찾든, 다른 분야에서 데이터를 다루든, 이 책의 목표는 선택한 주제 영역에 대해 새로운 질문을 하고 답할 수 있는 능력을 제공하는 것입니다.

## 이 책은 누구를 위한 것인가?

워싱턴 대학교와 다양한 기술 중심 컨퍼런스 및 밋업에서의 강의에서 제가 가장 많이 들었던 질문 중 하나는 다음과 같습니다: "파이썬을 어떻게 배워야 할까요?"
질문하는 사람들은 일반적으로 기술적으로 숙련된 학생, 개발자 또는 연구원이며, 종종 이미 코드 작성 및 계산 및 수치 도구 사용에 대한 강력한 배경 지식을 가지고 있습니다.
이들 중 대부분은 파이썬 자체를 배우고 싶어하는 것이 아니라, 데이터 집약적 및 계산 과학을 위한 도구로 사용하기 위해 언어를 배우고 싶어합니다.
이러한 청중을 위한 방대한 양의 비디오, 블로그 게시물 및 튜토리얼이 온라인에서 제공되지만, 저는 오랫동안 이 질문에 대한 단일한 좋은 답변이 없다는 점에 좌절해 왔습니다. 그것이 이 책을 쓰게 된 계기입니다.

이 책은 파이썬이나 프로그래밍 일반에 대한 소개를 위한 것이 아닙니다. 독자가 함수 정의, 변수 할당, 객체의 메서드 호출, 프로그램 흐름 제어 및 기타 기본 작업을 포함하여 파이썬 언어에 익숙하다고 가정합니다.
대신, 파이썬 사용자가 다음 섹션에서 언급된 라이브러리와 관련 도구와 같은 파이썬의 데이터 과학 스택을 사용하여 데이터를 효과적으로 저장, 조작하고 통찰력을 얻을 수 있도록 돕기 위한 것입니다.

## 왜 파이썬인가?

파이썬은 지난 수십 년 동안 대규모 데이터 세트의 분석 및 시각화를 포함한 과학 컴퓨팅 작업을 위한 최고 수준의 도구로 부상했습니다.
이는 파이썬 언어의 초기 지지자들에게는 놀라운 일이었을 수 있습니다: 언어 자체는 데이터 분석이나 과학 컴퓨팅을 염두에 두고 특별히 설계되지 않았습니다.
데이터 과학을 위한 파이썬의 유용성은 주로 다음과 같은 타사 패키지의 크고 활발한 생태계에서 비롯됩니다: 동종 배열 기반 데이터 조작을 위한 *NumPy*, 이기종 및 레이블이 지정된 데이터 조작을 위한 *Pandas*, 일반적인 과학 컴퓨팅 작업을 위한 *SciPy*, 출판 품질 시각화를 위한 *Matplotlib*, 대화형 실행 및 코드 공유를 위한 *IPython*, 기계 학습을 위한 *Scikit-Learn*, 그리고 다음 페이지에서 언급될 더 많은 도구들.

파이썬 언어 자체에 대한 가이드를 찾고 있다면 이 책의 자매 프로젝트인 [https://www.oreilly.com/library/view/a-whirlwind-tour/9781492037859](_A Whirlwind Tour of the Python Language_)를 제안합니다.
이 짧은 보고서는 이미 하나 이상의 다른 프로그래밍 언어에 익숙한 데이터 과학자를 대상으로 파이썬 언어의 필수 기능에 대한 둘러보기를 제공합니다.

## 책의 개요

이 책의 각 번호가 매겨진 부분은 파이썬 데이터 과학 이야기의 기본 부분을 구성하는 특정 패키지 또는 도구에 중점을 두며, 각 장은 단일 개념을 논의하는 짧은 독립적인 장으로 나뉩니다:

- *1부, Jupyter: 일반 파이썬을 넘어서*, IPython 및 Jupyter를 소개합니다. 이러한 패키지는 많은 파이썬 사용 데이터 과학자가 작업하는 계산 환경을 제공합니다.
- *2부, NumPy 소개*, NumPy 라이브러리에 중점을 둡니다. 이 라이브러리는 파이썬에서 밀집 데이터 배열의 효율적인 저장 및 조작을 위한 `ndarray`를 제공합니다.
- *3부, Pandas를 사용한 데이터 조작*, Pandas 라이브러리를 소개합니다. 이 라이브러리는 파이썬에서 레이블이 지정된/열 형식 데이터의 효율적인 저장 및 조작을 위한 `DataFrame`을 제공합니다.
- *4부, Matplotlib을 사용한 시각화*, Matplotlib에 집중합니다. 이 라이브러리는 파이썬에서 유연한 범위의 데이터 시각화 기능을 제공합니다.
- *5부, 기계 학습*, Scikit-Learn 라이브러리에 중점을 둡니다. 이 라이브러리는 가장 중요하고 확립된 기계 학습 알고리즘의 효율적이고 깔끔한 파이썬 구현을 제공합니다.

PyData 세계는 확실히 이 여섯 가지 패키지보다 훨씬 크며 매일 성장하고 있습니다.
이를 염두에 두고, 저는 이 책 전반에 걸쳐 파이썬에서 수행할 수 있는 작업의 경계를 넓히고 있는 다른 흥미로운 노력, 프로젝트 및 패키지에 대한 참조를 제공하기 위해 모든 노력을 기울입니다.
그럼에도 불구하고 제가 집중하는 패키지는 현재 파이썬 데이터 과학 분야에서 수행되는 많은 작업의 기본이며, 생태계가 계속 성장함에 따라 중요하게 유지될 것으로 예상합니다.

## 설치 고려 사항

파이썬과 과학 컴퓨팅을 가능하게 하는 라이브러리 제품군을 설치하는 것은 간단합니다. 이 섹션에서는 컴퓨터를 설정할 때 염두에 두어야 할 몇 가지 사항을 간략하게 설명합니다.

파이썬을 설치하는 다양한 방법이 있지만, 데이터 과학에서 사용하기 위해 제안하는 방법은 Anaconda 배포판이며, Windows, Linux 또는 macOS를 사용하든 유사하게 작동합니다.
Anaconda 배포판은 두 가지 버전으로 제공됩니다:

- [Miniconda](http://conda.pydata.org/miniconda.html)는 파이썬 인터프리터 자체와 함께 *conda*라는 명령줄 도구를 제공합니다. 이 도구는 Linux 사용자가 익숙할 수 있는 apt 또는 yum 도구와 유사한 정신으로 파이썬 패키지에 맞춰진 교차 플랫폼 패키지 관리자로 작동합니다.

- [Anaconda](https://www.continuum.io/downloads)에는 파이썬과 conda가 모두 포함되어 있으며, 추가적으로 과학 컴퓨팅에 맞춰진 다른 사전 설치된 패키지 제품군이 번들로 제공됩니다. 이 번들의 크기 때문에 설치 시 수 기가바이트의 디스크 공간이 소요될 것으로 예상됩니다.

Anaconda에 포함된 모든 패키지는 Miniconda 위에 수동으로 설치할 수도 있습니다. 이러한 이유로 Miniconda로 시작하는 것이 좋습니다.

시작하려면 Miniconda 패키지를 다운로드하여 설치하십시오(파이썬 3이 포함된 버전을 선택해야 함). 그런 다음 이 책에서 사용되는 핵심 패키지를 설치하십시오:

```
[~]$ conda install numpy pandas scikit-learn matplotlib seaborn jupyter
```

본문 전체에서 파이썬의 과학 생태계에 있는 다른 더 전문화된 도구도 사용할 것입니다. 설치는 일반적으로 **`conda install packagename`**을 입력하는 것만큼 쉽습니다.
기본 conda 채널에서 사용할 수 없는 패키지를 발견하면 광범위한 커뮤니티 기반 conda 패키지 저장소인 [*conda-forge*](https://conda-forge.org/)를 확인하십시오.

conda 환경 생성 및 사용에 대한 정보(강력히 권장)를 포함하여 conda에 대한 자세한 내용은 [conda의 온라인 설명서](http://conda.pydata.org/docs/)를 참조하십시오.
