이 책의 이 부분은 [3부](03.00-Introduction-to-Pandas.ipynb)와 함께 파이썬에서 메모리 내 데이터를 효과적으로 로드, 저장 및 조작하는 기술을 간략하게 설명합니다.
주제는 매우 광범위합니다. 데이터 세트는 문서 모음, 이미지 모음, 사운드 클립 모음, 수치 측정값 모음 또는 거의 모든 다른 것을 포함하여 광범위한 소스와 광범위한 형식으로 제공될 수 있습니다.
이러한 명백한 이질성에도 불구하고 많은 데이터 세트는 기본적으로 숫자 배열로 표현될 수 있습니다.

예를 들어 이미지, 특히 디지털 이미지는 영역 전체의 픽셀 밝기를 나타내는 단순한 2차원 숫자 배열로 생각할 수 있습니다.
사운드 클립은 시간에 따른 강도의 1차원 배열로 생각할 수 있습니다.
텍스트는 특정 단어 또는 단어 쌍의 빈도를 나타내는 이진 숫자와 같이 다양한 방식으로 숫자 표현으로 변환될 수 있습니다.
데이터가 무엇이든 분석 가능하게 만드는 첫 번째 단계는 숫자 배열로 변환하는 것입니다.
(이 프로세스의 몇 가지 구체적인 예는 [특성 공학](05.04-Feature-Engineering.ipynb)에서 논의할 것입니다.)

이러한 이유로 숫자 배열의 효율적인 저장 및 조작은 데이터 과학을 수행하는 프로세스에 절대적으로 기본입니다.
이제 이러한 숫자 배열을 처리하기 위한 파이썬의 특수 도구인 NumPy 패키지와 Pandas 패키지([3부](03.00-Introduction-to-Pandas.ipynb)에서 설명)를 살펴보겠습니다.

이 책의 이 부분에서는 NumPy를 자세히 다룹니다. NumPy(*Numerical Python*의 약자)는 밀집 데이터 버퍼를 저장하고 작동하기 위한 효율적인 인터페이스를 제공합니다.
어떤 면에서 NumPy 배열은 파이썬의 기본 제공 `list` 유형과 유사하지만 NumPy 배열은 배열 크기가 커짐에 따라 훨씬 효율적인 저장 및 데이터 작업을 제공합니다.
NumPy 배열은 파이썬의 거의 모든 데이터 과학 도구 생태계의 핵심을 형성하므로 NumPy를 효과적으로 사용하는 방법을 배우는 데 시간을 투자하면 어떤 데이터 과학 분야에 관심이 있든 가치가 있을 것입니다.

서문에서 설명한 조언을 따르고 Anaconda 스택을 설치했다면 이미 NumPy가 설치되어 바로 사용할 수 있습니다.
직접 설치하는 유형이라면 http://www.numpy.org/ 로 이동하여 거기에 있는 설치 지침을 따르십시오.
그런 다음 NumPy를 가져오고 버전을 다시 확인할 수 있습니다.

````python
import numpy
numpy.__version__
````

관례적으로 SciPy/PyData 세계의 대부분의 사람들은 NumPy를 `np`라는 별칭으로 가져올 것입니다.
