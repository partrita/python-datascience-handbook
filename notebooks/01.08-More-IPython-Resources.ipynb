# 더 많은 IPython 리소스

이 장들에서는 데이터 과학 작업을 가능하게 하기 위해 IPython을 사용하는 것에 대해 간략하게만 다루었습니다.
인쇄물과 웹 모두에서 훨씬 더 많은 정보를 얻을 수 있으며, 여기서는 도움이 될 수 있는 몇 가지 다른 리소스를 나열합니다.

## 웹 리소스

- [IPython 웹사이트](http://ipython.org): IPython 웹사이트는 설명서, 예제, 자습서 및 기타 다양한 리소스에 대한 링크를 제공합니다.
- [nbviewer 웹사이트](http://nbviewer.jupyter.org/): 이 사이트는 인터넷에서 사용할 수 있는 모든 Jupyter 노트북의 정적 렌더링을 보여줍니다. 첫 페이지에는 다른 사람들이 IPython을 무엇에 사용하는지 볼 수 있는 몇 가지 예제 노트북이 있습니다!
- [선별된 Jupyter 노트북 컬렉션](https://github.com/jupyter/jupyter/wiki): nbviewer로 구동되는 이 계속 성장하는 노트북 목록은 IPython으로 수행할 수 있는 수치 분석의 깊이와 폭을 보여줍니다. 짧은 예제와 자습서부터 노트북 형식으로 구성된 본격적인 과정과 책까지 모든 것을 포함합니다!
- 비디오 자습서: 인터넷을 검색하면 IPython에 대한 많은 비디오 자습서를 찾을 수 있습니다. 특히 IPython 및 Jupyter의 주요 제작자이자 유지 관리자인 페르난도 페레즈와 브라이언 그레인저가 PyCon, SciPy 및 PyData 컨퍼런스에서 제공하는 자습서를 찾는 것이 좋습니다.

## 도서

- [*Python for Data Analysis* (O'Reilly)](http://shop.oreilly.com/product/0636920023784.do): 웨스 맥키니의 이 책에는 데이터 과학자로서 IPython을 사용하는 방법을 다루는 장이 포함되어 있습니다. 많은 내용이 여기서 논의한 내용과 중복되지만 다른 관점은 항상 도움이 됩니다.
- [*Learning IPython for Interactive Computing and Data Visualization* (Packt)](https://www.packtpub.com/big-data-and-business-intelligence/learning-ipython-interactive-computing-and-data-visualization): 시릴 로산트의 이 짧은 책은 데이터 분석을 위해 IPython을 사용하는 방법에 대한 좋은 소개를 제공합니다.
- [*IPython Interactive Computing and Visualization Cookbook* (Packt)](https://www.packtpub.com/big-data-and-business-intelligence/ipython-interactive-computing-and-visualization-cookbook): 또한 시릴 로산트의 이 책은 데이터 과학을 위해 IPython을 사용하는 방법에 대한 더 길고 고급적인 내용을 다룹니다. 이름에도 불구하고 IPython에만 국한되지 않고 광범위한 데이터 과학 주제에 대해서도 심도 있게 다룹니다.

마지막으로, 스스로 도움을 찾을 수 있다는 점을 상기시켜 드립니다. IPython의 `?` 기반 도움말 기능([IPython 도움말 및 설명서](01.01-Help-And-Documentation.ipynb)에서 설명)은 잘 사용하고 자주 사용하면 유용할 수 있습니다.
여기 및 다른 곳의 예제를 살펴보면서 IPython이 제공하는 모든 도구에 익숙해질 수 있습니다.
