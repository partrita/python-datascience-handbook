# 입출력 내역

이전에 IPython 셸을 사용하면 위쪽 및 아래쪽 화살표 키 또는 동등한 Ctrl-p/Ctrl-n 단축키를 사용하여 이전 명령에 액세스할 수 있다는 것을 확인했습니다.
또한 셸과 노트북 모두에서 IPython은 이전 명령의 출력과 명령 자체의 문자열 버전을 얻는 여러 가지 방법을 제공합니다.
여기서는 이에 대해 살펴보겠습니다.

## IPython의 In 및 Out 객체

이제 여러분은 IPython에서 사용하는 `In [1]:`/`Out[1]:` 스타일의 프롬프트에 익숙해지셨을 것입니다.
그러나 이는 단지 예쁜 장식이 아니라는 것이 밝혀졌습니다. 이는 현재 세션에서 이전 입력 및 출력에 어떻게 액세스할 수 있는지에 대한 단서를 제공합니다.
다음과 같은 세션을 시작한다고 가정해 보겠습니다.

``ipython
[1]에서: 수학 가져오기

[2]에서: math.sin(2)
출력[2]: 0.9092974268256817

[3]에서: math.cos(2)
출력[3]: -0.4161468365471424
````

내장 `math` 패키지를 가져온 다음 숫자 2의 사인과 코사인을 계산했습니다.
이러한 입력 및 출력은 `In`/`Out` 레이블과 함께 셸에 표시되지만 더 많은 것이 있습니다. IPython은 실제로 이 기록을 반영하기 위해 자동으로 업데이트되는 `In` 및 `Out`이라는 파이썬(Python) 변수를 생성합니다.

``ipython
[4]에서: 에서
출력[4]: ['', '수학 가져오기', 'math.sin(2)', 'math.cos(2)', 'In']

입력 [5]: 출력
출력[5]:
{2: 0.9092974268256817,
3: -0.4161468365471424,
4: ['', '수학 가져오기', 'math.sin(2)', 'math.cos(2)', 'In', 'Out']}
````

`In` 개체는 명령을 순서대로 추적하는 목록입니다(목록의 첫 번째 항목은 `In [1]`이 첫 번째 명령을 참조할 수 있도록 자리 표시자입니다).

``ipython
[6]에서: 인쇄([1]에서)
수학 가져오기
````

`Out` 개체는 목록이 아니라 입력 번호를 출력(있는 경우)에 매핑하는 사전입니다.

``ipython
[7]에서: 인쇄(출력[2])
0.9092974268256817
````

모든 작업에 출력이 있는 것은 아닙니다. 예를 들어 `import` 문과 `print` 문은 출력에 영향을 주지 않습니다.
후자는 놀랄 수도 있지만 `print`가 `None`을 반환하는 함수라는 점을 고려하면 의미가 있습니다. 간결하게 하기 위해 'None'을 반환하는 명령은 'Out'에 추가되지 않습니다.

이것이 유용할 수 있는 곳은 과거 결과와 상호 작용하려는 경우입니다.
예를 들어 이전에 계산된 결과를 사용하여 `sin(2) ** 2`와 `cos(2) ** 2`의 합을 확인해 보겠습니다.

``ipython
입력 [8]: 출력[2] ** 2 + 출력[3] ** 2
출력[8]: 1.0
````

결과는 잘 알려진 삼각 항등식에서 예상한 대로 '1.0'입니다.
이 경우 이전 결과를 사용할 필요는 없지만 매우 비용이 많이 드는 계산을 실행하고 결과를 변수에 할당하는 것을 잊어버린 경우 매우 편리할 수 있습니다.

## 밑줄 단축키 및 이전 출력

표준 파이썬(Python) 셸에는 이전 출력에 액세스하기 위한 간단한 단축키가 하나만 포함되어 있습니다. 변수 '_'(즉, 단일 밑줄)는 이전 출력으로 계속 업데이트됩니다. 이는 IPython에서도 작동합니다:

``ipython
[9]에서: 인쇄(_)
1.0
````

그러나 IPython은 이를 조금 더 발전시킵니다. 이중 밑줄을 사용하여 마지막에서 두 번째 출력에 액세스하고 삼중 밑줄을 사용하여 마지막에서 세 번째 출력에 액세스할 수 있습니다(출력이 없는 명령은 건너뛰기).

``ipython
[10]에서: 인쇄(__)
-0.4161468365471424

[11]에서: 인쇄(___)
0.9092974268256817
````

IPython은 거기에서 멈춥니다. 3개 이상의 밑줄이 있으면 계산하기가 약간 어려워지기 시작하고, 그 시점에서 줄 번호로 출력을 참조하는 것이 더 쉽습니다.

하지만 제가 언급해야 할 단축키가 하나 더 있습니다. `Out[X]`의 약어는 `_X`입니다(즉, 단일 밑줄 뒤에 줄 번호가 옵니다).

``ipython
입력 [12]: 출력[2]
출력[12]: 0.9092974268256817

[13]에서: _2
출력[13]: 0.9092974268256817
````

## 출력 억제
때로는 명령문의 출력을 억제하고 싶을 수도 있습니다(이는 아마도 [Matplotlib 소개](04.00-Introduction-To-Matplotlib.ipynb)에서 살펴볼 플로팅 명령에서 가장 일반적일 것입니다).
또는 실행 중인 명령이 출력 기록에 저장하고 싶지 않은 결과를 생성하여 다른 참조가 제거될 때 할당이 취소될 수도 있습니다.
명령 출력을 억제하는 가장 쉬운 방법은 줄 끝에 세미콜론을 추가하는 것입니다.

``ipython
[14]에서: math.sin(2) + math.cos(2);
````

결과는 자동으로 계산되며 출력은 화면에 표시되거나 'Out' 사전에 저장되지 않습니다.

``ipython
입력 [15]: 14인치 출력
출력[15]: 거짓
````

## 관련 마법 명령
이전 입력 배치에 한 번에 액세스하려면 '%history' 매직 명령이 매우 유용합니다.
처음 4개의 입력을 인쇄하는 방법은 다음과 같습니다.

``ipython
[16]에서: %history -n 1-3
1: 수학 가져오기
2: math.sin(2)
3: math.cos(2)
````

평소와 같이 '%history?'를 입력하면 사용 가능한 옵션에 대한 자세한 정보와 설명을 볼 수 있습니다(``?' 기능에 대한 자세한 내용은 [IPython의 도움말과 문서](01.01-Help-And-Documentation.ipynb) 참조).
다른 유용한 마법 명령으로는 명령 기록의 일부를 다시 실행하는 '%rerun'과 명령 기록의 일부 세트를 파일에 저장하는 '%save'가 있습니다.
