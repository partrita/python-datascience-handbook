# IPython 및 셸 명령

표준 Python 인터프리터를 사용하여 대화형으로 작업할 때 좌절감 중 하나는 Python 도구와 시스템 명령줄 도구에 액세스하기 위해 여러 창 사이를 전환해야 한다는 것입니다.
IPython은 이러한 격차를 해소하고 IPython 터미널 내에서 직접 셸 명령을 실행하기 위한 구문을 제공합니다.
느낌표와 함께 마법이 일어납니다. 한 줄에서 `!` 뒤에 나타나는 모든 것은 Python 커널이 아니라 시스템 명령줄에 의해 실행됩니다.

다음 논의에서는 Linux 또는 macOS와 같은 Unix 계열 시스템을 사용하고 있다고 가정합니다.
다음 예제 중 일부는 기본적으로 다른 유형의 셸을 사용하는 Windows에서 실패합니다. 하지만 *Linux*용 Windows 하위 시스템을 사용하는 경우 여기 예제는 올바르게 실행되어야 합니다.
쉘 명령에 익숙하지 않다면 항상 훌륭한 Software Carpentry Foundation에서 제공하는 [Unix 쉘 튜토리얼](http://swcarpentry.github.io/shell-novice/)을 검토해 보시기 바랍니다.

## 셸에 대한 빠른 소개

쉘/터미널/명령줄 사용에 대한 전체 소개는 이 장의 범위를 훨씬 벗어나지만, 초보자를 위해 여기에서 간단한 소개를 제공하겠습니다.
쉘은 컴퓨터와 텍스트로 상호 작용하는 방법입니다.
1980년대 중반 Microsoft와 Apple이 이제 어디서나 볼 수 있는 그래픽 운영 체제의 첫 번째 버전을 출시한 이후 대부분의 컴퓨터 사용자는 친숙한 메뉴 선택과 끌어서 놓기 동작을 통해 운영 체제와 상호 작용해 왔습니다.
그러나 운영 체제는 이러한 그래픽 사용자 인터페이스 이전에 존재했으며 주로 일련의 텍스트 입력을 통해 제어되었습니다. 즉, 프롬프트에서 사용자가 명령을 입력하면 컴퓨터는 사용자가 지시한 대로 수행됩니다.
이러한 초기 프롬프트 시스템은 오늘날 대부분의 데이터 과학자가 여전히 사용하는 셸 및 터미널의 전신이었습니다.

셸에 익숙하지 않은 사람은 아이콘과 메뉴를 클릭하기만 하면 동일한 결과를 얻을 수 있는데 왜 이것을 귀찮게 하느냐고 물을 수 있습니다.
셸 사용자는 또 다른 질문으로 답할 수 있습니다. 입력을 통해 훨씬 더 쉽게 작업을 수행할 수 있는데 왜 아이콘과 메뉴 항목을 찾아야 할까요?
전형적인 기술 선호의 난관처럼 들릴 수도 있지만 기본 작업을 넘어서면 셸이 고급 작업에 대해 훨씬 더 많은 제어 기능을 제공한다는 사실이 금방 분명해집니다. 물론 학습 곡선이 위협적일 수는 있습니다.

예를 들어, 다음은 사용자가 시스템의 디렉터리와 파일을 탐색, 생성 및 수정하는 Linux/macOS 셸 세션의 샘플입니다(`osx:~ $`는 프롬프트이고 `$` 뒤의 모든 내용은 입력된 명령입니다. `#` 앞에 오는 텍스트는 실제로 입력하는 것이 아니라 설명을 의미합니다).

``배쉬
osx:~ $ echo "hello world" # echo는 Python의 인쇄 기능과 같습니다.
안녕 세계

osx:~ $ pwd # pwd = 작업 디렉토리 인쇄
/home/jake # 이것이 우리가 앉아 있는 "경로"입니다

osx:~ $ ls # ls = 작업 디렉터리 내용 나열
노트북 프로젝트

osx:~ $ cd projects/ # cd = 디렉토리 변경

osx:프로젝트 $ 비밀번호
/홈/제이크/프로젝트

osx:프로젝트 $ ls
datasci_book mpld3 myproject.txt

osx:projects $ mkdir myproject # mkdir = 새 디렉토리 만들기

osx:프로젝트 $ cd myproject/

osx:myproject $ mv ../myproject.txt ./ # mv = 파일 이동. 여기서 우리는
# 한 디렉터리의 myproject.txt 파일
# 현재 디렉터리(./)로 이동(../)합니다.
OSX:내 프로젝트 $ls
내프로젝트.txt
````

이 모든 것은 아이콘과 메뉴를 클릭하는 대신 명령을 입력하여 익숙한 작업(디렉토리 구조 탐색, 디렉토리 생성, 파일 이동 등)을 수행하는 간단한 방법일 뿐입니다.
몇 가지 명령(`pwd`, `ls`, `cd`, `mkdir` 및 `cp`)만으로 가장 일반적인 파일 작업을 대부분 수행할 수 있지만 이러한 기본 사항을 뛰어넘는 경우 셸 접근 방식이 정말 강력해집니다.

## IPython의 쉘 명령

모든 표준 쉘 명령은 앞에 `!` 문자를 붙여 IPython에서 직접 사용할 수 있습니다.
예를 들어 `ls`, `pwd` 및 `echo` 명령은 다음과 같이 실행할 수 있습니다.

``ipython
[1]에서: !ls
내프로젝트.txt

[2]에서: !pwd
/홈/제이크/프로젝트/myproject

[3]에서: !echo "쉘에서 인쇄"
쉘에서 인쇄하기
````

## 셸과 값 전달

셸 명령은 IPython에서 호출할 수 있을 뿐만 아니라 IPython 네임스페이스와 상호 작용하도록 만들 수도 있습니다.
예를 들어 할당 연산자 `=`를 사용하여 셸 명령의 출력을 Python 목록에 저장할 수 있습니다.

```ipython
In [4]: contents = !ls

In [5]: print(contents)
['myproject.txt']

In [6]: directory = !pwd

In [7]: print(directory)
['/Users/jakevdp/notebooks/tmp/myproject']
```

These results are not returned as lists, but as a special shell return type defined in IPython:

```ipython
In [8]: type(directory)
IPython.utils.text.SList
```

This looks and acts a lot like a Python list but has additional functionality, such as
the `grep` and `fields` methods and the `s`, `n`, and `p` properties that allow you to search, filter, and display the results in convenient ways.
For more information on these, you can use IPython's built-in help features.

Communication in the other direction—passing Python variables into the shell—is possible using the `{varname}` syntax:

```ipython
In [9]: message = "hello from Python"

In [10]: !echo {message}
hello from Python
```

The curly braces contain the variable name, which is replaced by the variable's contents in the shell command.

## Shell-Related Magic Commands

If you play with IPython's shell commands for a while, you might notice that you cannot use `!cd` to navigate the filesystem:

```ipython
In [11]: !pwd
/home/jake/projects/myproject

In [12]: !cd ..

In [13]: !pwd
/home/jake/projects/myproject
```

The reason is that shell commands in the notebook are executed in a temporary subshell that does not maintain state from command to command.
If you'd like to change the working directory in a more enduring way, you can use the `%cd` magic command:

```ipython
In [14]: %cd ..
/home/jake/projects
```

In fact, by default you can even use this without the `%` sign:

```ipython
In [15]: cd myproject
/home/jake/projects/myproject
```

This is known as an *automagic* function, and the ability to execute such commands without an explicit `%` can be toggled with the `%automagic` magic function.

Besides `%cd`, other available shell-like magic functions are `%cat`, `%cp`, `%env`, `%ls`, `%man`, `%mkdir`, `%more`, `%mv`, `%pwd`, `%rm`, and `%rmdir`, any of which can be used without the `%` sign if `automagic` is on.
This makes it so that you can almost treat the IPython prompt as if it's a normal shell:

```ipython
In [16]: mkdir tmp

In [17]: ls
myproject.txt  tmp/

In [18]: cp myproject.txt tmp/

In [19]: ls tmp
myproject.txt

In [20]: rm -r tmp
```

This access to the shell from within the same terminal window as your Python session lets you more naturally combine Python and the shell in your workflows with fewer context switches.
