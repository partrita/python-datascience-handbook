# I파이썬(Python) 및 셸 명령

표준 파이썬(Python) 인터프리터를 사용하여 대화형으로 작업할 때 좌절감 중 하나는 파이썬(Python) 도구와 시스템 커맨드 라인 도구에 액세스하기 위해 여러 창 사이를 전환해야 한다는 것입니다.
I파이썬(Python)은 이러한 격차를 해소하고 I파이썬(Python) 터미널 내에서 직접 셸 명령을 실행하기 위한 구문을 제공합니다.
느낌표와 함께 마법이 일어납니다. 한 줄에서 `!` 뒤에 나타나는 모든 것은 파이썬(Python) 커널이 아니라 시스템 커맨드 라인에 의해 실행됩니다.

다음 논의에서는 Linux 또는 macOS와 같은 Unix 계열 시스템을 사용하고 있다고 가정합니다.
다음 예제 중 일부는 기본적으로 다른 유형의 셸을 사용하는 Windows에서 실패합니다. 하지만 *Linux*용 Windows 하위 시스템을 사용하는 경우 여기 예제는 올바르게 실행되어야 합니다.
쉘 명령에 익숙하지 않다면 항상 훌륭한 Software Carpentry Foundation에서 제공하는 [Unix 쉘 튜토리얼](http://swcarpentry.github.io/shell-novice/)을 검토해 보시기 바랍니다.

## 셸에 대한 빠른 소개

쉘/터미널/커맨드 라인 사용에 대한 전체 소개는 이 장의 범위를 훨씬 벗어나지만, 초보자를 위해 여기에서 간단한 소개를 제공하겠습니다.
쉘은 컴퓨터와 텍스트로 상호 작용하는 방법입니다.
1980년대 중반 Microsoft와 Apple이 이제 어디서나 볼 수 있는 그래픽 운영 체제의 첫 번째 버전을 출시한 이후 대부분의 컴퓨터 사용자는 친숙한 메뉴 선택과 끌어서 놓기 동작을 통해 운영 체제와 상호 작용해 왔습니다.
그러나 운영 체제는 이러한 그래픽 사용자 인터페이스 이전에 존재했으며 주로 일련의 텍스트 입력을 통해 제어되었습니다. 즉, 프롬프트에서 사용자가 명령을 입력하면 컴퓨터는 사용자가 지시한 대로 수행됩니다.
이러한 초기 프롬프트 시스템은 오늘날 대부분의 데이터 과학(Data Science)자가 여전히 사용하는 셸 및 터미널의 전신이었습니다.

셸에 익숙하지 않은 사람은 아이콘과 메뉴를 클릭하기만 하면 동일한 결과를 얻을 수 있는데 왜 이것을 귀찮게 하느냐고 물을 수 있습니다.
셸 사용자는 또 다른 질문으로 답할 수 있습니다. 입력을 통해 훨씬 더 쉽게 작업을 수행할 수 있는데 왜 아이콘과 메뉴 항목을 찾아야 할까요?
전형적인 기술 선호의 난관처럼 들릴 수도 있지만 기본 작업을 넘어서면 셸이 고급 작업에 대해 훨씬 더 많은 제어 기능을 제공한다는 사실이 금방 분명해집니다. 물론 학습 곡선이 위협적일 수는 있습니다.

예를 들어 다음은 사용자가 시스템의 디렉터리와 파일을 탐색, 생성 및 수정하는 Linux/macOS 셸 세션의 샘플입니다(`osx:~ $`는 프롬프트이고 `$` 뒤의 모든 내용은 입력된 명령입니다. `#` 앞에 오는 텍스트는 실제로 입력하는 것이 아니라 설명을 의미합니다).

``배쉬
osx:~ $ echo "hello world" # echo는 파이썬(Python)의 인쇄 기능과 같습니다.
안녕 세계

osx:~ $ pwd # pwd = 작업 디렉토리 인쇄
/home/jake # 이것이 우리가 앉아 있는 "경로"입니다

osx:~ $ ls # ls = 작업 디렉터리 내용 나열
노트북 프로젝트

osx:~ $ cd projects/ # cd = 디렉토리 변경

osx:프로젝트 $ 비밀번호
/홈/제이크/프로젝트

osx:프로젝트 $ ls
datasci_book mpld3 myproject.txt

osx:projects $ mkdir myproject # mkdir = 새 디렉토리 만들기

osx:프로젝트 $ cd myproject/

osx:myproject $ mv ../myproject.txt ./ # mv = 파일 이동. 여기서 우리는
# 한 디렉터리의 myproject.txt 파일
# 현재 디렉터리(./)로 이동(../)합니다.
OSX:제 프로젝트 $ls
내프로젝트.txt
````

이 모든 것은 아이콘과 메뉴를 클릭하는 대신 명령을 입력하여 익숙한 작업(디렉토리 구조 탐색, 디렉토리 생성, 파일 이동 등)을 수행하는 간단한 방법일 뿐입니다.
몇 가지 명령(`pwd`, `ls`, `cd`, `mkdir` 및 `cp`)만으로 가장 일반적인 파일 작업을 대부분 수행할 수 있지만 이러한 기본 사항을 뛰어넘는 경우 셸 접근 방식이 정말 강력해집니다.

## I파이썬(Python)의 쉘 명령

모든 표준 쉘 명령은 앞에 `!` 문자를 붙여 I파이썬(Python)에서 직접 사용할 수 있습니다.
예를 들어 `ls`, `pwd` 및 `echo` 명령은 다음과 같이 실행할 수 있습니다.

``ipython
[1]에서: !ls
내프로젝트.txt

[2]에서: !pwd
/홈/제이크/프로젝트/myproject

[3]에서: !echo "쉘에서 인쇄"
쉘에서 인쇄하기
````

## 셸과 값 전달

셸 명령은 I파이썬(Python)에서 호출할 수 있을 뿐만 아니라 I파이썬(Python) 네임스페이스와 상호 작용하도록 만들 수도 있습니다.
예를 들어 할당 연산자 `=`를 사용하여 셸 명령의 출력을 파이썬(Python) 목록에 저장할 수 있습니다.

``ipython
[4]에서: 내용 = !ls

[5]에서: 인쇄(내용)
['myproject.txt']

[6]에서: 디렉토리 = !pwd

[7]에서: print(디렉토리)
['/Users/jakevdp/notebooks/tmp/myproject']
````

이러한 결과는 목록으로 반환되지 않고 I파이썬(Python)에 정의된 특수 셸 반환 유형으로 반환됩니다.

``ipython
[8]에서: 유형(디렉토리)
I파이썬(Python).utils.text.SList
````

이는 파이썬(Python) 목록과 매우 유사해 보이고 작동하지만 다음과 같은 추가 기능이 있습니다.
편리한 방법으로 결과를 검색, 필터링 및 표시할 수 있는 `grep` 및 `fields` 메소드와 `s`, `n` 및 `p` 속성이 있습니다.
이에 대한 자세한 내용을 보려면 I파이썬(Python)에 내장된 도움말 기능을 사용하세요.

파이썬(Python) 변수를 셸에 전달하는 반대 방향의 통신은 `{varname}` 구문을 사용하여 가능합니다.

``ipython
[9]에서: message = "hello from 파이썬(Python)"

[10]에서: !echo {메시지}
안녕하세요 파이썬에서
````

중괄호에는 변수 이름이 포함되어 있으며, 이는 쉘 명령의 변수 내용으로 대체됩니다.

## 쉘 관련 마법 명령

잠시 동안 I파이썬(Python)의 셸 명령을 사용하면 '!cd'를 사용하여 파일 시스템을 탐색할 수 없다는 것을 확인할 수 있습니다.

``ipython
[11]에서: !pwd
/홈/제이크/프로젝트/myproject

[12]에서: !cd ..

[13]에서: !pwd
/홈/제이크/프로젝트/myproject
````

그 이유는 노트북의 셸 명령은 명령마다 상태를 유지하지 않는 임시 하위 셸에서 실행되기 때문입니다.
보다 지속적인 방법으로 작업 디렉터리를 변경하려면 `%cd` 마법 명령을 사용할 수 있습니다.

``ipython
[14]에서: %cd ..
/홈/제이크/프로젝트
````

실제로 기본적으로 '%' 기호 없이도 이를 사용할 수 있습니다.

``ipython
[15]에서: cd myproject
/홈/제이크/프로젝트/myproject
````

이것은 *automagic* 기능으로 알려져 있으며 명시적인 `%` 없이 이러한 명령을 실행하는 기능은 `%automagic` 마법 기능으로 전환할 수 있습니다.

`%cd` 외에도 사용 가능한 다른 쉘형 매직 함수로는 `%cat`, `%cp`, `%env`, `%ls`, `%man`, `%mkdir`, `%more`, `%mv`, `%pwd`, `%rm`, `%rmdir`이 있으며, 이들 중 `automagic`이 켜져 있으면 `%` 기호 없이 사용할 수 있습니다.
이렇게 하면 I파이썬(Python) 프롬프트를 마치 일반 쉘인 것처럼 처리할 수 있습니다.

``ipython
[16]에서: mkdir tmp

[17]에서: ls
myproject.txt tmp/

[18]에서: cp myproject.txt tmp/

[19]에서: ls tmp
내프로젝트.txt

[20]에서: rm -r tmp
````

파이썬(Python) 세션과 동일한 터미널 창 내에서 셸에 액세스하면 더 적은 수의 컨텍스트 전환으로 워크플로에서 파이썬(Python)과 셸을 보다 자연스럽게 결합할 수 있습니다.
