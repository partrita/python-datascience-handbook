# 팬더를 이용한 데이터 조작

[2부](02.00-Introduction-to-NumPy.ipynb)에서는 NumPy와 파이썬(Python)에서 조밀한 형식의 배열을 효율적으로 저장하고 조작할 수 있는 'ndarray' 객체에 대해 자세히 살펴보았습니다.
여기서는 Pandas 라이브러리에서 제공하는 데이터 구조를 심층적으로 살펴봄으로써 이러한 지식을 구축해 보겠습니다.
Pandas는 `DataFrame`의 효율적인 구현을 제공하는 NumPy를 기반으로 구축된 최신 패키지입니다.
``DataFrame``은 본질적으로 행과 열 레이블이 첨부된 다차원 배열이며, 종종 이질적인 유형 및/또는 누락된 데이터가 있습니다.
Pandas는 레이블이 지정된 데이터를 위한 편리한 저장 인터페이스를 제공할 뿐만 아니라 데이터베이스 프레임워크와 스프레드시트 프로그램 사용자에게 친숙한 여러 가지 강력한 데이터 작업을 구현합니다.

우리가 본 것처럼 NumPy의 'ndarray' 데이터 구조는 수치 컴퓨팅 작업에서 일반적으로 볼 수 있는 깨끗하고 잘 구성된 데이터 유형에 필수적인 기능을 제공합니다.
이 목적은 매우 잘 수행되지만 더 많은 유연성이 필요할 때(예: 데이터에 레이블 첨부, 누락된 데이터 작업 등), 요소별 브로드캐스팅(예: 그룹화, 피벗 등)에 잘 매핑되지 않는 작업을 시도할 때 한계가 분명해집니다. 각 작업은 우리 주변의 다양한 형태로 제공되는 덜 구조화된 데이터를 분석하는 데 중요한 부분입니다.
Pandas, 특히 'Series' 및 'DataFrame' 개체는 NumPy 배열 구조를 기반으로 하며 데이터 과학(Data Science)자의 시간을 많이 차지하는 이러한 종류의 "데이터 정리" 작업에 대한 효율적인 액세스를 제공합니다.

책의 이 부분에서는 `Series`, `DataFrame` 및 관련 구조를 효과적으로 사용하는 메커니즘에 중점을 둘 것입니다.
적절한 경우 실제 데이터 세트에서 가져온 예를 사용하지만 이러한 예가 반드시 초점이 되는 것은 아닙니다.
