# Jupyter: 일반 Python을 넘어서

Python 개발 환경에는 다양한 옵션이 있으며, 내 작업에서 어떤 것을 사용하는지 묻는 경우가 많습니다.
내 대답은 때때로 사람들을 놀라게 합니다. 내가 선호하는 환경은 [IPython](http://ipython.org/)과 텍스트 편집기(내 경우에는 내 기분에 따라 Emacs 또는 VSCode)입니다.
Jupyter는 2001년 Fernando Perez가 향상된 Python 인터프리터로 만든 IPython 셸로 시작되었으며 이후 Perez의 표현에 따르면 "연구 컴퓨팅의 전체 수명 주기를 위한 도구"를 제공하는 것을 목표로 하는 프로젝트로 성장했습니다.
Python이 데이터 과학 작업의 엔진이라면 Jupyter를 대화형 제어판으로 생각할 수 있습니다.

Jupyter는 Python에 대한 유용한 대화형 인터페이스일 뿐만 아니라 언어에 여러 가지 유용한 구문 추가 기능도 제공합니다. 여기서는 이러한 추가 기능 중 가장 유용한 내용을 다루겠습니다.
Jupyter 프로젝트에서 제공하는 가장 친숙한 인터페이스는 아마도 데이터 과학 결과의 개발, 협업, 공유 및 게시에 유용한 브라우저 기반 환경인 Jupyter Notebook일 것입니다.
노트북 형식의 유용성을 보여주는 예로서, 읽고 있는 페이지만 살펴보세요. 이 책의 전체 원고는 Jupyter 노트북 세트로 구성되었습니다.

책의 이 부분은 데이터 과학 실습에 유용한 Jupyter 및 IPython 기능 중 일부를 단계별로 살펴보는 것으로 시작하며, 특히 Python의 표준 기능 이상으로 제공되는 구문에 중점을 둡니다.
다음으로, 데이터 과학 코드를 생성하고 사용하는 일반적인 작업 속도를 높일 수 있는 몇 가지 더 유용한 *마법 명령*에 대해 좀 더 깊이 살펴보겠습니다.
마지막으로 데이터를 이해하고 결과를 공유하는 데 유용한 노트북의 일부 기능을 살펴보겠습니다.
