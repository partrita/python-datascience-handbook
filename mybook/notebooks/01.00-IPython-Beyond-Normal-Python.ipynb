# Jupyter: 일반 파이썬(Python)을 넘어서

파이썬 개발 환경에는 다양한 옵션이 존재하며, 제게 어떤 환경을 사용하는지 묻는 이들이 많습니다.
제 대답은 때때로 사람들을 놀라게 합니다. 제가 가장 선호하는 환경은 [IPython](http://ipython.org/) 셸과 텍스트 에디터(제 경우 그날 기분에 따라 Emacs나 VSCode를 씁니다)이기 때문입니다.
Jupyter 프로젝트는 2001년 페르난도 페레즈(Fernando Perez)가 한층 더 강화된 파이썬 인터프리터를 목표로 만든 IPython 셸에서 비롯되었으며, 이후 페레즈의 표현처럼 '연구 컴퓨팅의 전체 생애주기를 위한 도구'를 제공하고자 하는 거대한 프로젝트로 성장했습니다.
파이썬이 데이터 과학(Data Science) 작업의 핵심 엔진이라면, Jupyter를 매끄러운 대화형 제어판이라고 생각할 수 있습니다.

Jupyter는 파이썬(Python)에 대한 유용한 대화형 인터페이스일 뿐만 아니라 언어에 여러 가지 유용한 구문 추가 기능도 제공합니다. 여기서는 이러한 추가 기능 중 가장 유용한 내용을 다루겠습니다.
Jupyter 프로젝트가 제공하는 가장 대중적인 인터페이스는 단연코 Jupyter Notebook일 것입니다. 이는 데이터 과학(Data Science) 결과물의 개발, 협업, 공유 및 게시에 매우 유용한 브라우저 기반 환경입니다.
노트북 포맷의 유용성을 직접 확인하고 싶다면 지금 읽고 계신 이 페이지만 살펴보셔도 충분합니다. 이 책의 전체 원고 자체가 여러 개의 Jupyter Notebook으로 작성되었습니다.

본 파트에서는 먼저 데이터 과학(Data Science) 실무에 유용한 다양한 Jupyter 및 IPython 기능들을 단계별로 살펴보고, 특히 파이썬의 표준 기능 이상으로 제공되는 유연한 구문들에 중점을 둡니다.
다음으로는 일반적인 데이터 과학(Data Science) 코드의 생성 및 활용 속도를 크게 높여주는 *매직 명령어(magic commands)*에 대해 훨씬 더 깊이 파고들어 봅니다.
마지막으로 데이터의 흐름을 빠르게 이해하고 결과를 세련되게 공유할 수 있도록 돕는 노트북의 다양한 부가 기능들을 둘러보겠습니다.
