# I파이썬(Python)과 Jupyter 시작하기

데이터 과학(Data Science)을 위한 파이썬 코드를 작성할 때 저는 보통 세 가지 작업 모드를 오가며 활용합니다: 간단하고 짧은 명령들을 빠르게 테스트할 때는 I파이썬(Python) 셸을, 결과를 누적하는 긴 대화형 분석과 콘텐츠 공유에는 Jupyter Notebook을, 그리고 다른 곳에서도 재사용이 가능한 파이썬 패키지를 만들 때는 Emacs나 VSCode와 같은 대화형 개발 환경(IDE)을 사용합니다.
따라서 이 파트에서는 앞서 언급한 두 가지 모드, 즉 I파이썬(Python) 셸과 Jupyter Notebook 환경에 온전히 집중해 보겠습니다.
소프트웨어 개발을 위해 전문적인 IDE를 다루는 것 역시 데이터 과학(Data Science)자의 필수 역량 중 하나지만, 이 책에서는 직접적으로 깊게 다루지는 않습니다.

## I파이썬(Python) 셸 시작하기

이 부분의 내용은, 책의 다른 부분과 마찬가지로, 단순히 눈으로만 읽고 넘어가도록 쓰이지 않았습니다.
코드를 읽는 동시에 여기서 다루는 다양한 도구와 구문을 직접 타이핑하고 실험해 볼 것을 강력히 권장합니다. 이렇게 키보드로 직접 쳐보며 형성된 '근육 기억'은 그저 눈으로만 문서를 읽는 행위보다 훨씬 더 가치 있고 유용하게 쓰일 것입니다.
커맨드 라인에 **`ipython`**을 입력하여 I파이썬(Python) 인터프리터를 시작합니다. 또는 앞서 언급한 Anaconda와 같은 배포판을 설치해서 사용 중인 경우, 운영체제별로 제공되는 특정 런처(launcher) 등을 통해 쉽게 실행하실 수 있습니다([I파이썬(Python) 도움말과 문서](01.01-Help-And-Documentation.ipynb) 참조).

## 주피터 노트북 실행하기

Jupyter Notebook은 근본적으로 I파이썬(Python) 셸을 브라우저 기반 그래픽 인터페이스로 구현한 것이며, 이를 기반으로 다양하고 풍부한 동적 디스플레이 기능들을 제공합니다. 일반적인 파이썬(파이썬(Python)) 및 I파이썬(Python) 코드를 실행하는 것을 넘어, 서식 있는 텍스트 문서, 정적 및 동적 데이터 시각화, 렌더링된 수학 수식, 그리고 JavaScript 대화형 위젯 등도 결과물에 유연하게 곁들일 수 있습니다. 또한, 이 노트북 포맷(`.ipynb`) 문서들은 다른 사람들이 문서를 열어 자신의 시스템에서 똑같이 코드를 실행하고 응용할 수 있는 형태로 아주 손쉽게 저장되고 공유됩니다.
