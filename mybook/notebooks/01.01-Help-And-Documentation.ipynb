통역사에 따라 이 정보는 인라인 텍스트로 표시되거나 별도의 팝업 창에 표시될 수 있습니다.

객체에 대한 도움말을 찾는 것은 매우 일반적이고 유용하기 때문에 IPython과 Jupyter는 이 문서 및 기타 관련 정보에 액세스하기 위한 약칭으로 '?' 문자를 도입합니다.

``ipython
[2]에서: len?
서명: len(obj, /)
Docstring: 컨테이너에 있는 항목 수를 반환합니다.
유형: 내장_함수_또는_방법
````

이 표기법은 객체 메서드를 포함한 거의 모든 것에 적용됩니다.

``ipython
[3]에서: L = [1, 2, 3]
[4]에서: L.insert?
서명: L.insert(index, object, /)
Docstring: 색인 앞에 객체를 삽입합니다.
유형: 내장_함수_또는_방법
````

또는 해당 유형의 문서를 포함하는 객체 자체도 있습니다.

``ipython
[5]에서: L?
유형: 목록
문자열 형식: [1, 2, 3]
길이: 3
독스트링:
내장된 가변 시퀀스.

인수가 제공되지 않으면 생성자는 새로운 빈 목록을 만듭니다.
지정된 경우 인수는 반복 가능해야 합니다.
````

중요한 것은 이 방법이 여러분이 직접 만든 함수나 다른 개체에도 적용된다는 것입니다!
여기에서는 독스트링을 사용하여 작은 함수를 정의하겠습니다:

``ipython
[6]에서: def square(a):
....: """a의 제곱을 반환합니다."""
....: ** 2를 반환합니다.
  ....:
````

함수에 대한 독스트링을 생성하려면 첫 번째 줄에 문자열 리터럴을 배치하면 됩니다.
독스트링은 일반적으로 여러 줄이므로 관례에 따라 여러 줄 문자열에 대해 Python의 삼중따옴표 표기법을 사용했습니다.

이제 `?`를 사용하여 이 독스트링을 찾습니다:

``ipython
[7]에서: 정사각형?
서명: 정사각형(a)
Docstring: a의 제곱을 반환합니다.
파일: <ipython-input-6>
유형: 기능
````

독스트링을 통해 문서에 빠르게 접근할 수 있다는 점은 여러분이 작성하는 코드에 항상 인라인 문서를 추가하는 습관을 들여야 하는 이유 중 하나입니다!

### ??로 소스 코드에 접근하기

Python 언어는 읽기가 매우 쉽기 때문에 일반적으로 궁금한 개체의 소스 코드를 읽으면 또 다른 수준의 통찰력을 얻을 수 있습니다.
IPython과 Jupyter는 이중 물음표(`??`)를 사용하여 소스 코드에 대한 바로 가기를 제공합니다.

``ipython
[8]에서: 정사각형??
서명: 정사각형(a)
원천:
데프 제곱(a):
"""a의 제곱을 반환합니다."""
** 2를 반환
파일: <ipython-input-6>
유형: 기능
````

이와 같은 간단한 기능의 경우 이중 물음표를 사용하면 내부 세부 정보에 대한 빠른 통찰력을 얻을 수 있습니다.

이것저것 많이 가지고 놀다 보면 때로는 `??` 접미사가 어떤 소스 코드도 표시하지 않는다는 것을 알게 될 것입니다. 이는 일반적으로 문제의 객체가 Python이 아니라 C 또는 다른 컴파일된 확장 언어로 구현되기 때문입니다.
이 경우 `??` 접미사는 `?` 접미사와 동일한 출력을 제공합니다.
특히 이전의 `len` 함수를 포함하여 Python의 많은 내장 객체 및 유형에서 이를 찾을 수 있습니다.

``ipython
[9]에서: len??
서명: len(obj, /)
Docstring: 컨테이너에 있는 항목 수를 반환합니다.
유형: 내장_함수_또는_방법
````

`?` 및/또는 `??`를 사용하는 것은 Python 함수나 모듈이 수행하는 작업에 대한 정보를 찾는 강력하고 빠른 방법입니다.

### 탭 완성 기능이 있는 모듈 탐색

또 다른 유용한 인터페이스는 개체, 모듈 및 네임스페이스의 내용을 자동 완성하고 탐색하기 위해 Tab 키를 사용하는 것입니다.
다음 예제에서는 '<TAB>'을 사용하여 Tab 키를 눌러야 하는 시점을 나타냅니다.

#### 개체 내용의 탭 완성

모든 Python 객체에는 이와 관련된 다양한 속성과 메서드가 있습니다.
앞서 언급한 `help` 함수와 마찬가지로 Python에는 이들 목록을 반환하는 `dir` 함수가 내장되어 있지만 실제로는 탭 완성 인터페이스가 훨씬 사용하기 쉽습니다.
개체의 사용 가능한 모든 속성 목록을 보려면 개체 이름 뒤에 마침표("`.`") 문자와 Tab 키를 입력하면 됩니다.

``ipython
[10]에서: L.<TAB>
추가() 카운트 삽입 역방향
확장 팝 정렬 지우기
복사 색인 제거
````

목록의 범위를 좁히려면 이름의 첫 번째 문자나 여러 문자를 입력하면 Tab 키를 눌러 일치하는 속성과 메서드를 찾을 수 있습니다.

``ipython
[10]에서: L.c<TAB>
클리어() 카운트()
복사()

[10]에서: L.co<TAB>
복사() 카운트()
````

옵션이 하나만 있는 경우 Tab 키를 누르면 줄이 완성됩니다.
예를 들어 다음은 `L.count`로 즉시 대체됩니다.

``ipython
[10]에서: L.cou<TAB>

````

Python에서는 공개/외부 속성과 비공개/내부 속성을 엄격하게 구분하지 않지만 관례적으로 앞의 밑줄을 사용하여 후자를 나타냅니다.
명확성을 위해 이러한 비공개 메서드와 특수 메서드는 기본적으로 목록에서 생략되지만 명시적으로 밑줄을 입력하여 나열할 수 있습니다.

``ipython
[10]에서: L._<TAB>
__추가__ __delattr__ __eq__
__class__ __delitem__ __format__()
__class_getitem__() __dir__() __ge__ >
__contains__ __doc__ __getattribute__
````

간결하게 하기 위해 출력의 처음 몇 열만 표시했습니다.
이들 중 대부분은 Python의 특별한 이중 밑줄 메서드(종종 "dunder" 메서드라는 별명이 붙음)입니다.

#### 가져오기 시 탭 완성

탭 완성은 패키지에서 개체를 가져올 때도 유용합니다.
여기서는 `co`로 시작하는 `itertools` 패키지에서 가능한 모든 가져오기를 찾는 데 이를 사용합니다.

``ipython
[10]: itertools import co<TAB>에서
조합() 압축()
Combinations_with_replacement() 개수()
````

마찬가지로, 탭 완성을 사용하여 시스템에서 어떤 가져오기를 사용할 수 있는지 확인할 수 있습니다(이는 Python 세션에 표시되는 타사 스크립트 및 모듈에 따라 변경됩니다).

``ipython

aifc 앱노페 >
반중력 아르곤2

`*` 문자는 빈 문자열을 포함한 모든 문자열과 일치합니다.

마찬가지로 이름 어딘가에 'find'라는 단어가 포함된 문자열 메서드를 찾고 있다고 가정해 보겠습니다.
다음과 같은 방법으로 검색할 수 있습니다.

``ipython
[11]에서: str.*find*?
str.find
str.rfind
````

이러한 유형의 유연한 와일드카드 검색은 새로운 패키지를 알아보거나 익숙한 패키지를 다시 알아볼 때 특정 명령을 찾는 데 유용할 수 있습니다.
