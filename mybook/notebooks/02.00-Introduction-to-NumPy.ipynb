# NumPy 소개

책의 이 부분은 [3부](03.00-Introduction-to-Pandas.ipynb)와 함께 파이썬(Python)에서 메모리 제 데이터를 효과적으로 로드, 저장 및 조작하는 기술을 간략하게 설명합니다.
주제는 매우 광범위합니다. 데이터 세트는 문서 컬렉션, 이미지 컬렉션, 사운드 클립 컬렉션, 수치 측정 컬렉션 또는 기타 거의 모든 것을 포함하여 다양한 소스와 형식으로 제공될 수 있습니다.
이러한 명백한 이질성에도 불구하고 많은 데이터세트는 기본적으로 숫자 배열로 표현될 수 있습니다.

예를 들어 이미지, 특히 디지털 이미지는 해당 영역의 픽셀 밝기를 나타내는 단순히 2차원 숫자 배열로 생각할 수 있습니다.
사운드 클립은 강도 대 시간의 1차원 배열로 생각할 수 있습니다.
텍스트는 특정 단어나 단어 쌍의 빈도를 나타내는 이진수와 같은 숫자 표현으로 다양한 방법으로 변환될 수 있습니다.
데이터가 무엇이든 분석 가능하게 만드는 첫 번째 단계는 데이터를 숫자 배열로 변환하는 것입니다.
(이 프로세스의 구체적인 예는 [Feature Engineering](05.04-Feature-Engineering.ipynb)에서 논의하겠습니다.)

이러한 이유로 숫자 배열의 효율적인 저장 및 조작은 데이터 과학(Data Science)을 수행하는 프로세스의 절대적 기본입니다.
이제 이러한 숫자 배열을 처리하기 위해 파이썬(Python)이 가지고 있는 특수 도구인 NumPy 패키지와 Pandas 패키지([3부](03.00-Introduction-to-Pandas.ipynb)에서 논의)를 살펴보겠습니다.

책의 이 부분에서는 NumPy를 자세히 다룰 것입니다. NumPy(*Numerical 파이썬(Python)*의 약어)는 밀도가 높은 데이터 버퍼를 저장하고 작동할 수 있는 효율적인 인터페이스를 제공합니다.
어떤 면에서 NumPy 배열은 파이썬(Python)에 내장된 '목록' 유형과 비슷하지만 NumPy 배열은 배열의 크기가 커짐에 따라 훨씬 더 효율적인 저장 및 데이터 작업을 제공합니다.
NumPy 배열은 파이썬(Python)의 거의 전체 데이터 과학(Data Science) 도구 생태계의 핵심을 형성하므로 NumPy를 효과적으로 사용하는 방법을 배우는 데 투자한 시간은 데이터 과학(Data Science)의 어떤 측면에 관심이 있든 상관없이 가치가 있을 것입니다.

서문에 설명된 조언에 따라 Anaconda 스택을 설치했다면 이미 NumPy가 설치되어 사용할 준비가 된 것입니다.
스스로 하는 것을 선호한다면 http://www.numpy.org/로 이동하여 거기에 있는 설치 지침을 따르십시오.
그런 다음 NumPy를 가져오고 버전을 다시 확인할 수 있습니다.

`````python
import numpy
numpy.__version__
`````

여기에서 논의된 패키지의 경우 NumPy 버전 1.8 이상을 권장합니다.
관례에 따르면 SciPy/PyData 세계의 대부분의 사람들은 'np'를 별칭으로 사용하여 NumPy를 가져옵니다.

`````python
import numpy as np
`````

이 장과 책의 나머지 부분 전체에서 이것이 우리가 NumPy를 가져오고 사용하는 방식이라는 것을 알게 될 것입니다.
