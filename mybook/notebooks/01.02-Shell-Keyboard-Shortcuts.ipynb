# I파이썬(Python) 셸의 키보드 단축키

컴퓨터에서 시간을 보내신다면 작업 흐름에서 키보드 단축키를 사용하신 적이 있을 것입니다.
아마도 가장 친숙한 것은 다양한 프로그램과 시스템에서 복사하고 붙여넣는 데 사용되는 Cmd-c와 Cmd-v(또는 Ctrl-c와 Ctrl-v)일 것입니다.
고급 사용자는 훨씬 더 나아가는 경향이 있습니다. Emacs, Vim 등과 같은 널리 사용되는 텍스트 편집기는 복잡한 키 입력 조합을 통해 사용자에게 놀라운 작업 범위를 제공합니다.

I파이썬(Python) 셸은 여기까지는 아니지만 명령을 입력하는 동안 빠른 탐색을 위한 다양한 키보드 단축키를 제공합니다.
이러한 바로 가기 중 일부는 브라우저 기반 노트북에서 작동하지만 이번 섹션에서는 주로 I파이썬(Python) 셸의 바로 가기에 대해 설명합니다.

이러한 기능에 익숙해지면 "홈" 키보드 위치에서 손을 움직이지 않고도 특정 명령을 빠르게 수행하는 데 매우 유용할 수 있습니다.
Emacs 사용자이거나 Linux 스타일 셸에 대한 경험이 있다면 다음 내용이 매우 익숙할 것입니다.
이러한 단축키를 *탐색 단축키*, *텍스트 입력 단축키*, *명령 기록 단축키*, *기타 단축키* 등 몇 가지 카테고리로 그룹화하겠습니다.

## 탐색 단축키

줄에서 앞뒤로 이동하기 위해 왼쪽 및 오른쪽 화살표 키를 사용하는 것은 매우 분명하지만 "홈" 키보드 위치에서 손을 움직일 필요가 없는 다른 옵션도 있습니다.

| 키스트로크 | 액션 |
|---------|------------------|
| Ctrl-a | 커서를 줄의 시작 부분으로 이동 |
| Ctrl-e | 커서를 줄 끝으로 이동 |
| Ctrl-b 또는 왼쪽 화살표 키 | 커서를 한 문자 뒤로 이동 |
| Ctrl-f 또는 오른쪽 화살표 키 | 커서를 한 문자 앞으로 이동 |

## 텍스트 입력 단축키

이전 문자를 삭제하기 위해 백스페이스 키를 사용하는 것은 누구나 익숙하지만 키를 누르려면 약간의 손가락 연습이 필요한 경우가 많으며 한 번에 문자 하나만 삭제됩니다.
I파이썬(Python)에는 입력 중인 텍스트의 일부를 제거할 수 있는 몇 가지 단축키가 있습니다. 이 중 가장 즉각적으로 유용한 것은 전체 텍스트 줄을 삭제하는 명령입니다.
이전 문자를 삭제하기 위해 백스페이스를 누르는 대신 Ctrl-b와 Ctrl-d를 조합하여 사용하는 경우 이것이 제2의 천성이 되었다는 것을 알게 될 것입니다!

| 키스트로크 | 액션 |
|---|------------|
| 백스페이스 키 | 줄의 이전 문자 삭제 |
| Ctrl-d | 줄의 다음 문자 삭제 |
| Ctrl-k | 커서부터 줄 끝까지 텍스트 자르기 |
| Ctrl-u | 줄의 시작 부분부터 커서까지 텍스트 자르기 |
| Ctrl-y | 이전에 잘라낸 텍스트를 양크(즉, 붙여넣기) |
| Ctrl-t | 이전 두 문자 바꾸기(즉, 전환) |

## 명령 기록 단축키

아마도 여기서 논의된 가장 영향력 있는 단축키는 I파이썬(Python)이 명령 기록 탐색을 위해 제공하는 단축키일 것입니다.
이 명령 기록은 현재 I파이썬(Python) 세션 이상입니다. 전체 명령 기록은 I파이썬(Python) 프로필 디렉터리의 SQLite 데이터베이스에 저장됩니다.
이전 명령에 액세스하는 가장 간단한 방법은 위쪽 및 아래쪽 화살표 키를 사용하여 기록을 단계별로 살펴보는 것이지만 다른 옵션도 있습니다.

| 키스트로크 | 액션 |
|---------|------------------|
| Ctrl-p(또는 위쪽 화살표 키) | 기록에서 이전 명령에 액세스 |
| Ctrl-n(또는 아래쪽 화살표 키) | 기록에서 다음 명령에 액세스 |
| Ctrl-r | 명령 내역을 통한 역검색 |

역방향 검색 옵션이 특히 유용할 수 있습니다.
앞서 우리가 `square`라는 함수를 정의했던 것을 기억하세요.
새로운 I파이썬(Python) 셸에서 파이썬(Python) 기록을 역검색하여 이 정의를 다시 찾아보겠습니다.
I파이썬(Python) 터미널에서 Ctrl-r을 누르면 다음 프롬프트가 표시됩니다.

``ipython
[1]에서:
(역방향 검색)`':
````

이 프롬프트에서 문자를 입력하기 시작하면 I파이썬(Python)은 해당 문자와 ​​일치하는 가장 최근 명령을 자동으로 채웁니다.

``ipython
[1]에서:
(역방향 검색)`sqa': 정사각형??
````

언제든지 더 많은 문자를 추가하여 검색을 구체화하거나 Ctrl-r을 다시 눌러 쿼리와 일치하는 다른 명령을 추가로 검색할 수 있습니다. 앞서 따라한 경우 Ctrl-r을 두 번 더 누르면 다음이 제공됩니다.

``ipython
[1]에서:
(역방향 검색)`sqa': def square(a):
"""a의 제곱을 반환합니다"""
** 2를 반환
````

찾고 있는 명령을 찾은 후 Return 키를 누르면 검색이 종료됩니다.
그런 다음 검색된 명령을 사용하고 세션을 계속할 수 있습니다.

``ipython
[1]에서: def square(a):
"""a의 제곱을 반환합니다"""
** 2를 반환

[2]에서: 정사각형(2)
아웃[2]: 4
````

Ctrl-p/Ctrl-n 또는 위쪽/아래쪽 화살표 키를 사용하여 비슷한 방식으로 기록을 검색할 수 있지만 줄 시작 부분의 문자를 일치시켜야만 검색할 수 있습니다.
즉, **`def`**를 입력한 다음 Ctrl-p를 누르면 기록에서 `def` 문자로 시작하는 가장 최근 명령(있는 경우)을 찾습니다.

## 기타 단축키

마지막으로, 이전 범주에 속하지 않지만 알아두면 유용한 몇 가지 기타 단축키가 있습니다.

| 키스트로크 | 액션 |
|------|------------------|
| Ctrl-l | 터미널 화면 지우기 |
| Ctrl-c | 현재 파이썬(Python) 명령 중단 |
| Ctrl-d | I파이썬(Python) 세션 종료 |

특히 Ctrl-c 단축키는 매우 오래 실행되는 작업을 실수로 시작할 때 유용할 수 있습니다.

여기에서 설명하는 일부 단축키는 처음에는 다소 모호해 보일 수 있지만 연습하면 빠르게 자동으로 적용됩니다.
일단 근육 기억이 발달하면, 다른 상황에서도 그 근육 기억이 가능했으면 좋겠다는 생각을 하게 될 것입니다.
