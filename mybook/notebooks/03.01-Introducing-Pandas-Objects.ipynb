# 판다스 객체 소개

매우 기본적인 수준에서 Pandas 개체는 행과 열이 단순한 정수 인덱스가 아닌 레이블로 식별되는 NumPy 구조 배열의 향상된 버전으로 생각할 수 있습니다.
이 장의 과정에서 살펴보겠지만 Pandas는 기본 데이터 구조 외에 유용한 도구, 방법 및 기능을 제공하지만 다음의 거의 모든 내용에서는 이러한 구조가 무엇인지 이해해야 합니다.
따라서 더 진행하기 전에 세 가지 기본 Pandas 데이터 구조인 'Series', 'DataFrame' 및 'Index'를 살펴보겠습니다.

표준 NumPy 및 Pandas 가져오기로 코드 세션을 시작합니다.

`````python
import numpy as np
import pandas as pd
`````

## 팬더 시리즈 객체

Pandas '시리즈'는 인덱스 데이터의 1차원 배열입니다.
다음과 같이 목록이나 배열에서 만들 수 있습니다.

`````python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
data
`````

'Series'는 일련의 값을 명시적인 색인 순서와 결합하며, 이는 'values' 및 'index' 속성을 통해 액세스할 수 있습니다.
'값'은 단순히 친숙한 NumPy 배열입니다.

`````python
data.values
`````

`index`는 `pd.Index` 유형의 배열형 객체입니다. 이에 대해 잠시 더 자세히 설명하겠습니다.

`````python
data.index
`````

NumPy 배열과 마찬가지로, 익숙한 파이썬(Python) 대괄호 표기법을 통해 관련 인덱스를 통해 데이터에 액세스할 수 있습니다.

`````python
data[1]
`````

`````python
data[1:3]
`````

하지만 앞으로 살펴보겠지만 Pandas '시리즈'는 에뮬레이트하는 1차원 NumPy 배열보다 훨씬 더 일반적이고 유연합니다.

### 일반화된 NumPy 배열로서의 시리즈

지금까지 살펴본 것에서 `Series` 객체는 기본적으로 1차원 NumPy 배열과 상호 교환 가능한 것처럼 보일 수 있습니다.
근본적인 차이점은 NumPy 배열에는 값에 액세스하는 데 사용되는 *암시적으로 정의된* 정수 인덱스가 있는 반면, Pandas `Series`에는 값과 관련된 *명시적으로 정의된* 인덱스가 있다는 것입니다.

이 명시적인 인덱스 정의는 `Series` 객체에 추가 기능을 제공합니다. 예를 들어 인덱스는 정수일 필요는 없지만 원하는 유형의 값으로 구성될 수 있습니다.
따라서 원한다면 문자열을 인덱스로 사용할 수 있습니다.

`````python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
`````

그리고 항목 액세스는 예상대로 작동합니다.

`````python
data['b']
`````

비연속적이거나 비순차적 인덱스를 사용할 수도 있습니다.

`````python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=[2, 5, 3, 7])
data
`````

`````python
data[5]
`````

전문사전으로서의 ### 시리즈

이런 식으로 Pandas '시리즈'를 파이썬(Python) 사전의 전문화와 비슷하다고 생각할 수 있습니다.
사전은 임의의 키를 임의의 값 집합에 매핑하는 구조이고, '시리즈'는 입력된 키를 입력된 값 집합에 매핑하는 구조입니다.
이러한 입력은 중요합니다. NumPy 배열 뒤의 유형별 컴파일 코드가 특정 작업에 대해 파이썬(Python) 목록보다 더 효율적이듯이, Pandas `Series`의 유형 정보는 특정 작업에 대해 파이썬(Python) 사전보다 더 효율적입니다.

사전으로서의 'Series' 비유는 파이썬(Python) 사전에서 직접 'Series' 객체를 구성함으로써 더욱 명확해질 수 있습니다. 여기에는 2020년 인구 조사에 따르면 미국에서 가장 인구가 많은 5개 주가 있습니다.

`````python
population_dict = {'California': 39538223, 'Texas': 29145505,
                   'Florida': 21538187, 'New York': 20201249,
                   'Pennsylvania': 13002700}
population = pd.Series(population_dict)
population
`````

여기에서 일반적인 사전 스타일 항목 액세스를 수행할 수 있습니다.

`````python
population['California']
`````

그러나 사전과 달리 `Series`는 슬라이싱과 같은 배열 스타일 작업도 지원합니다.

`````python
population['California':'Florida']
`````

[데이터 인덱싱 및 선택](03.02-Data-Indexing-and-Selection.ipynb)에서 Pandas 인덱싱 및 슬라이싱의 몇 가지 단점에 대해 논의하겠습니다.

### 시리즈 객체 구성하기

우리는 이미 Pandas '시리즈'를 처음부터 구성하는 몇 가지 방법을 살펴보았습니다. 이들 모두는 다음의 일부 버전입니다.

``파이썬
pd.Series(데이터, 인덱스=인덱스)
````

여기서 `index`는 선택적 인수이고 `data`는 여러 엔터티 중 하나일 수 있습니다.

예를 들어 `data`는 목록이나 NumPy 배열일 수 있으며, 이 경우 `index`는 기본적으로 정수 시퀀스로 설정됩니다.

`````python
pd.Series([2, 4, 6])
`````

또는 `data`는 지정된 인덱스를 채우기 위해 반복되는 스칼라일 수 있습니다.

`````python
pd.Series(5, index=[100, 200, 300])
`````

또는 사전일 수도 있습니다. 이 경우 `index`는 기본적으로 사전 키로 설정됩니다.

`````python
pd.Series({2:'a', 1:'b', 3:'c'})
`````

각 경우에 인덱스는 사용된 키의 하위 집합이나 순서를 제어하도록 명시적으로 설정할 수 있습니다.

`````python
pd.Series({2:'a', 1:'b', 3:'c'}, index=[1, 2])
`````

## 팬더 DataFrame 객체

Pandas의 다음 기본 구조는 'DataFrame'입니다.
이전 섹션에서 설명한 `Series` 객체와 마찬가지로 `DataFrame`은 NumPy 배열의 일반화 또는 파이썬(Python) 사전의 특수화로 생각할 수 있습니다.
이제 이러한 관점을 각각 살펴보겠습니다.

### 일반화된 NumPy 배열로서의 DataFrame
'Series'가 명시적인 인덱스가 있는 1차원 배열과 유사하다면 'DataFrame'은 명시적인 행 및 열 인덱스가 있는 2차원 배열과 유사합니다.
2차원 배열을 정렬된 1차원 열의 순서화된 시퀀스로 생각하는 것처럼 'DataFrame'은 정렬된 'Series' 개체의 시퀀스로 생각할 수 있습니다.
여기서 "정렬"이란 동일한 인덱스를 공유한다는 의미입니다.

이를 입증하기 위해 먼저 이전 섹션에서 논의한 5개 주 각각의 면적(제곱 킬로미터)을 나열하는 새로운 '계열'을 구성해 보겠습니다.

`````python
area_dict = {'California': 423967, 'Texas': 695662, 'Florida': 170312, 
             'New York': 141297, 'Pennsylvania': 119280}
area = pd.Series(area_dict)
area
`````

이제 이전의 `population` 시리즈와 함께 이것을 얻었으므로 사전을 사용하여 이 정보를 포함하는 단일 2차원 개체를 구성할 수 있습니다.

`````python
states = pd.DataFrame({'population': population,
                       'area': area})
states
`````

`Series` 객체와 마찬가지로 `DataFrame`에는 인덱스 레이블에 대한 액세스를 제공하는 `index` 속성이 있습니다.

`````python
states.index
`````

또한 `DataFrame`에는 열 레이블을 보유하는 `Index` 객체인 `columns` 속성이 있습니다.

`````python
states.columns
`````

따라서 `DataFrame`은 행과 열 모두 데이터에 액세스하기 위한 일반화된 인덱스를 갖는 2차원 NumPy 배열의 일반화로 생각할 수 있습니다.

### 특수 사전으로서의 DataFrame

마찬가지로 `DataFrame`을 사전의 특수화로 생각할 수도 있습니다.
사전이 키를 값에 매핑하는 경우 `DataFrame`은 열 이름을 열 데이터의 `시리즈`에 매핑합니다.
예를 들어 'area' 속성을 요청하면 앞서 본 영역이 포함된 'Series' 개체가 반환됩니다.

`````python
states['area']
`````

여기서 혼동될 수 있는 점을 주목하세요. 2차원 NumPy 배열에서 `data[0]`은 첫 번째 *행*을 반환합니다. `DataFrame`의 경우 `data['col0']`는 첫 번째 *열*을 반환합니다.
이 때문에 ``DataFrame``을 일반화된 배열보다는 일반화된 사전으로 생각하는 것이 더 나을 것입니다. 하지만 상황을 보는 두 가지 방법 모두 유용할 수 있습니다.
[데이터 인덱싱 및 선택](03.02-Data-Indexing-and-Selection.ipynb)에서 ``DataFrame``을 인덱싱하는 보다 유연한 방법을 살펴보겠습니다.

### DataFrame 객체 구성

Pandas `DataFrame`은 다양한 방법으로 구성될 수 있습니다.
여기에서는 몇 가지 예를 살펴보겠습니다.

#### 단일 Series 객체에서

`DataFrame`은 `Series` 객체의 컬렉션이며, 단일 `Series`에서 단일 열 `DataFrame`을 구성할 수 있습니다.

`````python
pd.DataFrame(population, columns=['population'])
`````

#### 사전 목록에서

모든 사전 목록을 `DataFrame`으로 만들 수 있습니다.
우리는 간단한 목록 이해를 사용하여 일부 데이터를 생성할 것입니다:

`````python
data = [{'a': i, 'b': 2 * i}
        for i in range(3)]
pd.DataFrame(data)
`````

사전의 일부 키가 누락된 경우에도 Pandas는 이를 `NaN` 값으로 채웁니다(예: "숫자가 아님", [누락된 데이터 처리](03.04-Missing-Values.ipynb) 참조).

`````python
pd.DataFrame([{'a': 1, 'b': 2}, {'b': 3, 'c': 4}])
`````

#### Series 객체 사전에서

이전에 살펴본 것처럼 `DataFrame`은 `Series` 객체의 사전에서도 구성될 수 있습니다.

`````python
pd.DataFrame({'population': population,
              'area': area})
`````

#### 2차원 NumPy 배열에서

2차원 데이터 배열이 주어지면 지정된 열과 인덱스 이름을 사용하여 'DataFrame'을 만들 수 있습니다.
생략하면 정수 인덱스가 각각에 사용됩니다.

`````python
pd.DataFrame(np.random.rand(3, 2),
             columns=['foo', 'bar'],
             index=['a', 'b', 'c'])
`````

#### NumPy 구조 배열에서

[구조화된 데이터: NumPy의 구조화된 배열](02.09-Structured-Data-NumPy.ipynb)에서 구조화된 배열을 다루었습니다.
Pandas `DataFrame`은 구조화된 배열과 매우 유사하게 작동하며 다음 중 하나에서 직접 생성될 수 있습니다.

`````python
A = np.zeros(3, dtype=[('A', 'i8'), ('B', 'f8')])
A
`````

`````python
pd.DataFrame(A)
`````

## 팬더 인덱스 개체

보시다시피 `Series` 및 `DataFrame` 개체에는 모두 데이터를 참조하고 수정할 수 있는 명시적인 *인덱스*가 포함되어 있습니다.
이 'Index' 개체는 그 자체로 흥미로운 구조이며 *불변 배열* 또는 *순서가 지정된 집합*(기술적으로는 다중 집합, 'Index' 개체에 반복된 값이 포함될 수 있음)으로 생각할 수 있습니다.
이러한 뷰는 'Index' 개체에 사용할 수 있는 작업 측면에서 몇 가지 흥미로운 결과를 가져옵니다.
간단한 예로 정수 목록에서 `Index`를 구성해 보겠습니다.

`````python
ind = pd.Index([2, 3, 5, 7, 11])
ind
`````

### 불변 배열로서의 인덱스

여러 면에서 '인덱스'는 배열처럼 작동합니다.
예를 들어 표준 파이썬(Python) 인덱싱 표기법을 사용하여 값이나 조각을 검색할 수 있습니다.

`````python
ind[1]
`````

`````python
ind[::2]
`````

`Index` 객체에는 NumPy 배열과 친숙한 많은 속성도 있습니다.

`````python
print(ind.size, ind.shape, ind.ndim, ind.dtype)
`````

`Index` 객체와 NumPy 배열의 한 가지 차이점은 색인이 불변이라는 것입니다. 즉, 일반적인 수단을 통해 수정할 수 없습니다.

`````python
ind[1] = 0
`````

이러한 불변성은 부주의한 인덱스 수정으로 인한 부작용 가능성 없이 여러 ``DataFrame``과 배열 간에 인덱스를 공유하는 것을 더 안전하게 만듭니다.

### 주문 세트로서의 인덱스

Pandas 개체는 집합 산술의 여러 측면에 의존하는 데이터 세트 전체의 조인과 같은 작업을 용이하게 하도록 설계되었습니다.
`Index` 객체는 파이썬(Python)의 내장 `set` 데이터 구조에서 사용되는 많은 규칙을 따르므로 합집합, 교집합, 차이 및 기타 조합을 익숙한 방식으로 계산할 수 있습니다.

`````python
indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])
`````

`````python
indA.intersection(indB)
`````

`````python
indA.union(indB)
`````

`````python
indA.symmetric_difference(indB)
`````
