# 추가 리소스

책의 이 부분에서는 데이터 분석에 Pandas를 효과적으로 사용하는 데 필요한 많은 기본 사항을 다루었습니다.
그럼에도 불구하고 우리의 논의에서는 많은 부분이 생략되었습니다.
Pandas에 대해 자세히 알아보려면 다음 리소스를 추천합니다.

- [Pandas 온라인 문서](http://pandas.pydata.org/): 패키지의 전체 문서를 볼 수 있는 소스입니다. 문서의 예는 소규모로 생성된 데이터 세트를 기반으로 하는 경향이 있지만 옵션에 대한 설명은 완전하며 일반적으로 다양한 기능의 사용을 이해하는 데 매우 유용합니다.

- [*데이터 분석을 위한 파이썬(Python)*](https://learning.oreilly.com/library/view/python-for-data/9781098104023/): Wes McKinney(Pandas의 최초 창시자)가 쓴 이 책에는 이 장에서 설명할 수 있는 것보다 Pandas 패키지에 대해 훨씬 더 많은 세부 정보가 포함되어 있습니다. 특히 McKinney는 금융 컨설턴트로서 그의 기본이었던 시계열 도구에 대해 자세히 설명합니다. 이 책에는 실제 데이터 세트에서 통찰력을 얻기 위해 Pandas를 적용하는 흥미로운 사례도 많이 있습니다.

- [*Effective Pandas*](https://leanpub.com/active-pandas): Pandas 개발자 Tom Augspurger가 작성한 이 짧은 전자책은 Pandas 라이브러리의 모든 기능을 효과적이고 관용적인 방식으로 사용하는 방법에 대한 간결한 개요를 제공합니다.

- [PyVideo의 Pandas](http://pyvideo.org/search?q=pandas): PyCon에서 SciPy, PyData에 이르기까지 많은 컨퍼런스에서 Pandas 개발자와 고급 사용자가 제공하는 튜토리얼을 선보였습니다. 특히 PyCon 튜토리얼은 매우 잘 검증된 발표자가 제공하는 경향이 있습니다.

이러한 리소스를 이 장에 제공된 연습과 결합하여 사용하면 Pandas를 사용하여 직면하는 모든 데이터 분석 문제를 해결할 수 있기를 바랍니다!
