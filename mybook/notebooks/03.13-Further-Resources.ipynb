# 더 읽을거리

지금까지 데이터 분석을 위해 Pandas를 효과적으로 사용하는 데 필요한 핵심 기초 지식을 살펴보았습니다.
하지만 지면 관계상 다루지 못한 내용도 많습니다.
Pandas를 더 깊이 있게 공부하고 싶은 분들께 다음 리소스를 추천합니다.

- [Pandas 온라인 문서](http://pandas.pydata.org/): Pandas 패키지의 모든 문서를 확인할 수 있는 공식 사이트입니다. 예제 코드가 소규모 데이터셋 위주인 경향이 있지만, 각 옵션에 대한 설명이 상세하여 기능 전반을 이해하는 데 큰 도움이 됩니다.

- [*파이썬 라이브러리를 활용한 데이터 분석(Python for Data Analysis)*](https://learning.oreilly.com/library/view/python-for-data/9781098104023/): Pandas의 창시자인 웨스 맥키니(Wes McKinney)가 쓴 책으로, 이 책에서 다룬 것보다 훨씬 방대한 Pandas 활용법을 담고 있습니다. 특히 금융권 컨설턴트 경험을 바탕으로 한 시계열 도구 설명이 매우 탁월합니다. 실제 데이터셋을 활용해 인사이트를 도출하는 흥미로운 사례도 가득합니다.

- [*Effective Pandas*](https://leanpub.com/active-pandas): Pandas 개발자인 톰 어그스퍼거(Tom Augspurger)가 쓴 짧은 전자책입니다. Pandas의 다양한 기능을 효과적이고 파이썬답게(idiomatic) 사용하는 법을 간결하게 정리했습니다.

- [PyVideo의 Pandas](http://pyvideo.org/search?q=pandas): SciPy, PyData, PyCon 등 여러 컨퍼런스에서 발표된 Pandas 개발자와 숙련자들의 튜토리얼 영상을 모아둔 곳입니다. 특히 PyCon 튜토리얼은 검증된 연사들의 수준 높은 강의가 많아 추천할 만합니다.

여기서 소개한 리소스들을 이 책의 실습과 병행한다면, 어떤 데이터 분석 문제라도 Pandas를 이용해 거뜬히 해결할 수 있을 것입니다!
