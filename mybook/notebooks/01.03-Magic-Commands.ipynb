# IPython 매직 명령

이전 장에서는 IPython을 사용하여 파이썬(Python)을 효율적이고 대화형으로 사용하고 탐색하는 방법을 보여주었습니다.
여기서는 IPython이 일반 파이썬(Python) 구문 위에 추가하는 몇 가지 향상된 기능에 대해 논의하기 시작합니다.
이는 IPython에서 *마법 명령*으로 알려져 있으며 `%` 문자가 앞에 붙습니다.
이러한 마법 명령은 표준 데이터 분석에서 흔히 발생하는 다양한 문제를 간결하게 해결하도록 설계되었습니다.
매직 명령은 두 가지 형태로 제공됩니다. *라인 매직은 단일 `%` 접두사로 표시되고 단일 입력 라인에서 작동하며 *셀 매직*은 이중 `%%` 접두사로 표시되고 여러 입력 라인에서 작동합니다.
여기서는 몇 가지 간단한 예를 보여주고 논의하고 나중에 몇 가지 유용한 마법 명령에 대해 좀 더 집중적으로 논의하겠습니다.

## 외부 코드 실행: %run
보다 광범위한 코드 개발을 시작하면 대화형 탐색을 위해 IPython에서 작업하고 재사용하려는 코드를 저장하는 텍스트 편집기를 사용하게 될 것입니다.
이 코드를 새 창에서 실행하는 것보다 IPython 세션 내에서 실행하는 것이 편리할 수 있습니다.
이는 `%run` 마법 명령을 사용하여 수행할 수 있습니다.

예를 들어 다음 내용으로 *myscript.py* 파일을 만들었다고 가정해 보겠습니다.

``파이썬
# 파일: myscript.py

데프 제곱(x):
"""숫자를 제곱하세요"""
x ** 2 반환

범위(1, 4)의 N에 대해:
print(f"{N} 제곱은 {square(N)}")
````

다음과 같이 IPython 세션에서 이를 실행할 수 있습니다.

``ipython
[6]에서: %run myscript.py
1의 제곱은 1이다
2의 제곱은 4
3의 제곱은 9
````

또한 이 스크립트를 실행한 후에는 그 안에 정의된 모든 함수를 IPython 세션에서 사용할 수 있습니다.

``ipython
[7]에서: 정사각형(5)
아웃[7]: 25
````

코드 실행 방법을 세부적으로 조정할 수 있는 몇 가지 옵션이 있습니다. IPython 인터프리터에 **`%run?`**을 입력하면 일반적인 방법으로 문서를 살펴볼 수 있습니다.

## 타이밍 코드 실행: %timeit
유용한 마법 함수의 또 다른 예는 '%timeit'입니다. 이 함수는 뒤에 오는 한 줄 파이썬(Python) 문의 실행 시간을 자동으로 결정합니다.
예를 들어 목록 이해의 성능을 확인하고 싶을 수 있습니다.

``ipython
[8]에서: %timeit L = [n ** 2 for n in range(1000)]
루프당 430 µs ± 3.21 µs(7회 실행의 평균 ± 표준 편차, 각각 1000 루프)
````

`%timeit`의 이점은 짧은 명령의 경우 보다 강력한 결과를 얻기 위해 자동으로 여러 실행을 수행한다는 것입니다.
여러 줄 명령문의 경우 두 번째 '%' 기호를 추가하면 여러 줄의 입력을 처리할 수 있는 셀 매직으로 바뀔 것입니다.
예를 들어 다음은 `for` 루프를 사용한 동등한 구성입니다.

``ipython
[9]에서: %%timeit
...: L = []
...: 범위(1000)에 있는 n의 경우:
...: L.append(n ** 2)
   ...: 
루프당 484 µs ± 5.67 µs(7회 실행의 평균 ± 표준 편차, 각각 1000루프)
````

이 경우 목록 이해가 동등한 'for' 루프 구성보다 약 10% 더 빠르다는 것을 즉시 확인할 수 있습니다.
[프로파일링 및 타이밍 코드](01.07-Timing-and-Profiling.ipynb)에서 `%timeit`과 타이밍 및 프로파일링 코드에 대한 기타 접근 방식을 살펴보겠습니다.

## 매직 함수에 대한 도움말: ?, %magic 및 %lsmagic

일반 파이썬(Python) 함수와 마찬가지로 IPython 매직 함수에는 독스트링이 있으며 이는 유용합니다.
문서는 표준 방식으로 액세스할 수 있습니다.
예를 들어 `%timeit` 마법 함수의 문서를 읽으려면 다음과 같이 입력하면 됩니다.

``ipython
[10]에서: %timeit?
````

다른 기능에 대한 문서도 비슷하게 액세스할 수 있습니다.
몇 가지 예를 포함하여 사용 가능한 매직 기능에 대한 일반적인 설명에 액세스하려면 다음을 입력하십시오.

``ipython
[11]에서: %magic
````

사용 가능한 모든 마법 기능의 빠르고 간단한 목록을 보려면 다음을 입력하세요.

``ipython
[12]에서: %lsmagic
````

마지막으로, 원한다면 자신만의 마법 함수를 정의하는 것이 매우 간단하다는 점을 언급하겠습니다.
