# 머신러닝

이 마지막 부분은 주로 파이썬(Python)의 [Scikit-Learn](http://scikit-learn.org) 패키지를 통해 매우 광범위한 머신러닝(Machine Learning) 주제를 소개합니다.
머신러닝(Machine Learning)은 프로그램이 데이터 세트에서 특정 패턴을 감지하고 데이터로부터 "학습"하여 추론을 도출할 수 있도록 하는 알고리즘 클래스로 생각할 수 있습니다.
이는 머신러닝(Machine Learning) 분야에 대한 포괄적인 소개를 의미하지 않습니다. 이는 큰 주제이며 여기에서 취하는 것보다 더 기술적인 접근 방식이 필요합니다.
또한 Scikit-Learn 패키지 사용을 위한 포괄적인 매뉴얼도 아닙니다(이에 대해서는 [추가 머신러닝(Machine Learning) 리소스](05.15-Learning-More.ipynb)에 나열된 리소스를 참조할 수 있습니다).
오히려 여기서의 목표는 다음과 같습니다.

- 머신러닝의 기본 어휘와 개념을 소개합니다.
- Scikit-Learn API를 소개하고 사용 사례를 보여줍니다.
- 보다 중요한 몇 가지 기존 머신러닝(Machine Learning) 접근 방식에 대해 자세히 알아보고 작동 방식과 적용 시기 및 위치에 대한 직관을 개발합니다.

이 자료의 대부분은 제가 PyCon, SciPy, PyData 및 기타 컨퍼런스에서 여러 차례 제공한 Scikit-Learn 튜토리얼과 워크숍에서 가져온 것입니다.
