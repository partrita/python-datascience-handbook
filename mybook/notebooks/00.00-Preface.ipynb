# 서문

## 데이터 과학(Data Science)이란 무엇입니까?

이 책은 파이썬(파이썬(Python))을 활용한 데이터 과학(Data Science)을 다루며, 필연적으로 '*데이터 과학(Data Science)*이 과연 무엇인가?'라는 질문을 던지게 됩니다.
데이터 과학(Data Science)이라는 용어가 얼마나 흔하게 쓰이는지를 생각해보면, 그 정의를 내리기가 놀라울 정도로 어렵다는 사실은 꽤 당혹스럽습니다.
비평가들은 이를 불필요한 라벨(애초에 데이터를 다루지 않는 과학이 어디 있겠습니까?)이거나, 그저 이력서를 화려하게 포장하여 채용 담당자의 눈길을 끌기 위한 유행어에 불과하다고 일축하기도 합니다.

하지만 제 생각에 이런 비판은 중요한 핵심을 놓치고 있습니다.
과장된 측면이 있음을 감안하더라도, 데이터 과학(Data Science)은 산업계와 학계의 활용 분야 전반에 걸쳐 그 중요성이 커지고 있는 융합적(학제간) 기술 집합을 일컫는 가장 적절한 표현일 것입니다.
여기서 *융합적(학제간)* 이라는 점이 핵심입니다. 개인적으로, 데이터 과학(Data Science)에 대한 현존하는 가장 훌륭한 정의는 드류 콘웨이(Drew Conway)가 2010년 9월 자신의 블로그에 처음 소개한 데이터 과학(Data Science) 벤 다이어그램(Venn diagram)에 잘 나타나 있다고 생각합니다(아래 그림 참조).

![Data Science Venn Diagram](images/Data_Science_VD.png)

<small>(source: [Drew Conway](http://drewconway.com/zia/2013/3/26/the-data-science-venn-diagram), used by permission)</small>

교차로 레이블 중 일부는 다소 농담 같지만, 이 다이어그램은 사람들이 "데이터 과학(Data Science)"이라고 말할 때 의미하는 바의 본질을 포착합니다. 즉, 근본적으로 학제간 주제입니다.
데이터 과학(Data Science)은 세 가지 뚜렷하고 중복되는 영역으로 구성됩니다. 데이터 세트(점점 더 커지고 있음)를 모델링하고 요약하는 방법을 아는 *통계학자*의 기술; 이 데이터를 효율적으로 저장, 처리 및 시각화하기 위한 알고리즘을 설계하고 사용할 수 있는 *컴퓨터 과학자*의 기술; 그리고 *도메인 전문 지식*(우리가 특정 주제에 대한 "고전적인" 훈련이라고 생각할 수 있는 것)은 올바른 질문을 공식화하고 답변을 맥락에 맞추는 데 필요합니다.

이를 염두에 두고 데이터 과학(Data Science)을 배워야 할 새로운 지식 영역이 아니라 현재 전문 분야 내에서 적용할 수 있는 새로운 기술 세트로 생각하도록 권장합니다.
선거 결과 보고, 주식 수익률 예측, 온라인 광고 클릭 최적화, 현미경 사진에서 미생물 식별, 새로운 종류의 천체 탐색, 기타 분야의 데이터 작업 등 무엇을 하든 이 책의 목표는 선택한 주제 영역에 대해 새로운 질문을 하고 답할 수 있는 능력을 제공하는 것입니다.

## 이 책은 누구를 위한 책인가요?

워싱턴 대학교와 다양한 기술 중심 컨퍼런스 및 모임에서 제가 들었던 가장 일반적인 질문 중 하나는 "파이썬(Python)을 어떻게 배워야 하나요?"였습니다.
질문하는 사람들은 일반적으로 기술적인 생각을 갖고 있는 학생, 개발자 또는 연구원이며, 종종 이미 코드 작성과 계산 및 수치 도구 사용에 대한 강력한 배경 지식을 갖추고 있습니다.
이들 대부분은 파이썬(Python) 자체를 배우고 싶어하는 것이 아니라 파이썬(Python)을 데이터 집약적이고 계산 과학을 위한 도구로 사용하려는 목적으로 언어를 배우고 싶어합니다.
이 청중을 위한 대규모 비디오, 블로그 게시물 및 튜토리얼이 온라인에서 제공되지만 저는 이 질문에 대한 단 하나의 좋은 답변이 부족하여 오랫동안 좌절감을 느꼈습니다. 그것이 이 책에 영감을 준 것입니다.

이 책은 파이썬(Python)이나 일반적인 프로그래밍에 대한 소개가 아닙니다. 저는 독자가 함수 정의, 변수 할당, 객체의 메서드 호출, 프로그램 흐름 제어 및 기타 기본 작업을 포함하여 파이썬(Python) 언어에 익숙하다고 가정합니다.
대신 파이썬(Python) 사용자가 파이썬(Python)의 데이터 과학(Data Science) 스택(다음 섹션에서 언급한 라이브러리 및 관련 도구)을 사용하여 데이터를 효과적으로 저장, 조작하고 통찰력을 얻는 방법을 배울 수 있도록 돕기 위한 것입니다.

## 왜 파이썬인가?

파이썬(Python)은 지난 수십 년 동안 대규모 데이터 세트의 분석 및 시각화를 포함하여 과학 컴퓨팅 작업을 위한 최고의 도구로 등장했습니다.
이는 파이썬(Python) 언어의 초기 지지자들에게는 놀라운 일이었을 수 있습니다. 언어 자체는 데이터 분석이나 과학적 컴퓨팅을 염두에 두고 특별히 설계되지 않았습니다.
데이터 과학(Data Science)을 위한 파이썬(Python)의 유용성은 주로 동종 배열 기반 데이터 조작을 위한 *NumPy*, 이종 및 레이블이 지정된 데이터 조작을 위한 *Pandas*, 일반적인 과학 컴퓨팅 작업을 위한 *SciPy*, 출판 품질 시각화를 위한 *Matplotlib*, 코드 대화형 실행 및 공유를 위한 *IPython*, 머신러닝(Machine Learning)을 위한 *Scikit-Learn* 및 다음에 언급될 더 많은 도구 등 대규모의 활동적인 타사 패키지 생태계에서 비롯됩니다. 페이지.

파이썬(Python) 언어 자체에 대한 가이드를 찾고 있다면 이 책의 자매 프로젝트인 [https://www.oreilly.com/library/view/a-whirlwind-tour/9781492037859](_A Whirlwind Tour of the 파이썬(Python) Language_)를 추천합니다.
이 짧은 보고서는 이미 하나 이상의 다른 프로그래밍 언어에 익숙한 데이터 과학(Data Science)자를 대상으로 파이썬(Python) 언어의 필수 기능에 대한 둘러보기를 제공합니다.

## 책 개요

이 책의 번호가 매겨진 각 부분은 파이썬(Python) 데이터 과학(Data Science) 이야기의 기본 부분에 기여하는 특정 패키지 또는 도구에 중점을 두고 있으며, 각각 단일 개념을 논의하는 짧은 독립 장으로 나누어져 있습니다.

- *1부, Jupyter: 일반 파이썬(Python)을 넘어서*에서는 IPython과 Jupyter를 소개합니다. 이러한 패키지는 파이썬(Python)을 사용하는 많은 데이터 과학(Data Science)자가 작업하는 컴퓨팅 환경을 제공합니다.
- *2부, NumPy 소개*에서는 파이썬(Python)에서 밀도가 높은 데이터 배열을 효율적으로 저장하고 조작하기 위한 'ndarray'를 제공하는 NumPy 라이브러리에 중점을 둡니다.
- *3부, Pandas를 사용한 데이터 조작*에서는 파이썬(Python)에서 레이블이 지정된/열 데이터를 효율적으로 저장하고 조작하기 위한 'DataFrame'을 제공하는 Pandas 라이브러리를 소개합니다.
- *제4부, Matplotlib를 사용한 시각화*에서는 파이썬(Python)에서 유연한 범위의 데이터 시각화 기능을 제공하는 라이브러리인 Matplotlib에 중점을 둡니다.
- *파트 V, 머신러닝(Machine Learning)*은 가장 중요하고 확립된 머신러닝(Machine Learning) 알고리즘의 효율적이고 깔끔한 파이썬(Python) 구현을 제공하는 Scikit-Learn 라이브러리에 중점을 둡니다.

PyData 세계는 확실히 이 6개 패키지보다 훨씬 크며 매일 성장하고 있습니다.
이를 염두에 두고 저는 이 책 전반에 걸쳐 파이썬(Python)에서 수행할 수 있는 작업의 경계를 넓히는 다른 흥미로운 노력, 프로젝트 및 패키지에 대한 참조를 제공하기 위해 모든 노력을 기울였습니다.
그럼에도 불구하고, 제가 집중하고 있는 패키지는 현재 파이썬(Python) 데이터 과학(Data Science) 공간에서 수행되는 많은 작업의 기본이며, 생태계가 계속해서 성장하더라도 계속 중요할 것으로 기대합니다.

## 설치 고려 사항

과학적 컴퓨팅을 가능하게 하는 파이썬(Python) 및 라이브러리 제품군을 설치하는 것은 간단합니다. 이번 섹션에서는 컴퓨터를 설정할 때 염두에 두어야 할 몇 가지 사항에 대해 설명합니다.

파이썬(Python)을 설치하는 방법은 다양하지만 데이터 과학(Data Science)에 사용하기 위해 제가 제안하는 방법은 Windows, Linux 또는 macOS를 사용하든 유사하게 작동하는 Anaconda 배포판입니다.
Anaconda 배포판은 두 가지 형태로 제공됩니다.

- [Miniconda](http://conda.pydata.org/miniconda.html)는 Linux 사용자에게 익숙할 수 있는 apt 또는 yum 도구와 유사하게 파이썬(Python) 패키지에 맞춰진 크로스 플랫폼 패키지 관리자로 작동하는 *conda*라는 커맨드 라인 도구와 함께 파이썬(Python) 인터프리터 자체를 제공합니다.

- [Anaconda](https://www.continuum.io/downloads)에는 파이썬(Python)과 conda가 모두 포함되어 있으며 과학 컴퓨팅에 맞춰 사전 설치된 다른 패키지 제품군도 추가로 번들로 제공됩니다. 이 번들의 크기로 인해 설치 시 몇 기가바이트의 디스크 공간이 소모될 것으로 예상됩니다.

Anaconda에 포함된 모든 패키지는 Miniconda 위에 수동으로 설치할 수도 있습니다. 이런 이유로 저는 Miniconda로 시작하는 것을 제안합니다.

시작하려면 Miniconda 패키지를 다운로드하여 설치하고(파이썬(Python) 3 버전을 선택해야 함) 이 책에 사용된 핵심 패키지를 설치하십시오.

````
[~]$ conda install numpy pandas scikit-learn matplotlib seaborn jupyter
````

본문 전반에 걸쳐 파이썬(Python)의 과학 생태계에서 보다 전문화된 다른 도구도 사용할 것입니다. 설치는 일반적으로 **`conda install packagename`**을 입력하는 것만큼 쉽습니다.
기본 conda 채널에서 사용할 수 없는 패키지를 발견한 경우 커뮤니티 중심의 광범위한 conda 패키지 저장소인 [*conda-forge*](https://conda-forge.org/)를 확인하세요.

Conda 환경 생성 및 사용에 대한 정보(*강력히* 권장함)를 포함하여 Conda에 대한 자세한 내용은 [conda 온라인 설명서](http://conda.pydata.org/docs/)를 참조하세요.
