# 추가 IPython 리소스

이번 장에서는 IPython을 데이터 과학 작업에 활용하는 방법을 아주 일부분만 맛보았습니다.
웹상에는 물론이고 시중의 도서들에도 훨씬 더 많은 정보가 담겨 있습니다. 도움이 될 만한 몇 가지 리소스를 아래에 정리했습니다.

## 웹 리소스

- [IPython 웹사이트](http://ipython.org): 공식 문서를 비롯해 각종 예제와 튜토리얼 등 풍부한 자료를 제공합니다.
- [nbviewer 웹사이트](http://nbviewer.jupyter.org/): 인터넷상의 모든 주피터 노트북을 정적으로 렌더링해 보여주는 사이트입니다. 첫 페이지에 있는 여러 예제 노트북을 통해 다른 사람들이 IPython을 어떻게 활용하는지 엿볼 수 있습니다.
- [선별된 주피터 노트북 컬렉션](https://github.com/jupyter/jupyter/wiki): 계속 업데이트되는 이 목록은 IPython으로 할 수 있는 수치 분석의 깊이와 폭을 잘 보여줍니다. 간단한 예제부터 전문적인 강좌와 책 한 권 분량의 내용까지 아주 다양합니다.
- 비디오 튜토리얼: 인터넷 검색을 통해 IPython 관련 영상 자료를 많이 찾을 수 있습니다. 특히 IPython과 Jupyter의 창시자이자 핵심 개발자인 페르난도 페레즈(Fernando Perez)와 브라이언 그레인저(Brian Granger)가 진행한 PyCon, SciPy, PyData 컨퍼런스의 튜토리얼 영상을 추천합니다.

## 도서

- [*파이썬 라이브러리를 활용한 데이터 분석*(O'Reilly)](http://shop.oreilly.com/product/0636920023784.do): 웨스 맥키니(Wes McKinney)의 이 책에는 데이터 과학자로서 IPython을 활용하는 방법을 다룬 장이 포함되어 있습니다. 여기서 다룬 내용과 겹치는 부분도 많지만, 또 다른 시각에서 접근해 보는 것도 큰 도움이 됩니다.
- [*Learning IPython for Interactive Computing and Data Visualization*(Packt)](https://www.packtpub.com/big-data-and-business-intelligence/learning-ipython-interactive-computing-and-data-visualization): 시릴 로산트(Cyrille Rossant)가 쓴 이 책은 데이터 분석을 위한 IPython 활용법을 친절하게 소개합니다.
- [*IPython Interactive Computing and Visualization Cookbook*(Packt)](https://www.packtpub.com/big-data-and-business-intelligence/ipython-interactive-computing-and-visualization-cookbook): 역시 시릴 로산트가 쓴 책으로, 데이터 과학을 위한 IPython 활용법을 더 깊이 있게 다룹니다. 제목과는 달리 단순히 IPython에만 머무르지 않고, 폭넓은 데이터 과학 주제들을 심도 있게 파고듭니다.

마지막으로, 궁금한 점은 언제든 스스로 찾아볼 수 있다는 사실을 잊지 마세요. IPython의 `?` 기반 도움말 기능을 습관처럼 활용하면 큰 도움이 될 것입니다.
여기서 소개한 리소스와 다양한 예제들을 직접 따라 하다 보면, 어느새 IPython이 제공하는 강력한 도구들에 익숙해진 자신을 발견하게 될 것입니다.
