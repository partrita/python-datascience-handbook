# 추가 I파이썬(Python) 리소스

이 장 세트에서 우리는 I파이썬(Python)을 사용하여 데이터 과학(Data Science) 작업을 활성화하는 방법을 표면적으로만 살펴보았습니다.
훨씬 더 많은 정보가 인쇄물과 웹에서 제공됩니다. 여기서는 도움이 될 수 있는 기타 리소스를 나열하겠습니다.

## 웹 리소스

- [I파이썬(Python) 웹사이트](http://ipython.org): I파이썬(Python) 웹사이트는 문서, 예제, 튜토리얼 및 기타 다양한 리소스에 대한 링크를 제공합니다.
- [nbviewer 웹사이트](http://nbviewer.jupyter.org/): 이 사이트는 인터넷에서 사용할 수 있는 모든 Jupyter 노트북의 정적 렌더링을 보여줍니다. 첫 페이지에는 다른 사람들이 I파이썬(Python)을 어떤 용도로 사용하고 있는지 찾아볼 수 있는 몇 가지 예제 노트북이 있습니다!
- [Jupyter 노트북 선별 컬렉션](https://github.com/jupyter/jupyter/wiki): nbviewer에서 제공하는 계속 늘어나는 이 노트북 목록은 I파이썬(Python)으로 수행할 수 있는 수치 분석의 깊이와 폭을 보여줍니다. 여기에는 짧은 예제와 튜토리얼부터 본격적인 코스와 노트북 형식으로 구성된 책까지 모든 것이 포함되어 있습니다!
- 비디오 튜토리얼: 인터넷을 검색하면 I파이썬(Python)에 대한 많은 비디오 튜토리얼을 찾을 수 있습니다. 특히 I파이썬(Python)과 Jupyter의 주요 창시자이자 유지관리자인 Fernando Perez와 Brian Granger가 진행하는 PyCon, SciPy 및 PyData 컨퍼런스의 튜토리얼을 찾아보는 것을 추천합니다.

## 도서

- [*데이터 분석을 위한 파이썬(Python)*(O'Reilly)](http://shop.oreilly.com/product/0636920023784.do): Wes McKinney의 책에는 데이터 과학(Data Science)자로서 I파이썬(Python)을 사용하는 방법을 다루는 장이 포함되어 있습니다. 많은 자료가 여기서 논의한 내용과 중복되지만, 다른 관점이 항상 도움이 됩니다.
- [*대화형 컴퓨팅 및 데이터 시각화를 위한 I파이썬(Python) 학습*(Packt)](https://www.packtpub.com/big-data-and-business-intelligence/learning-ipython-interactive-computing-and-data-visualization): Cyrille Rossant가 쓴 이 짧은 책은 데이터 분석을 위한 I파이썬(Python) 사용에 대한 좋은 소개를 제공합니다.
- [*I파이썬(Python) 대화형 컴퓨팅 및 시각화 요리책*(Packt)](https://www.packtpub.com/big-data-and-business-intelligence/ipython-interactive-computing-and-visualization-cookbook): 또한 Cyrille Rossant가 쓴 이 책은 데이터 과학(Data Science)을 위한 I파이썬(Python) 사용에 대한 더 길고 고급 처리입니다. 이름에도 불구하고 이는 단지 I파이썬(Python)에 관한 것이 아닙니다. 또한 광범위한 데이터 과학(Data Science) 주제에 대해 어느 정도 깊이 있게 설명합니다.

마지막으로, 스스로 도움말을 찾을 수 있다는 점을 상기시켜 드립니다. I파이썬(Python)의 `?` 기반 도움말 기능([I파이썬(Python)의 도움말과 문서](01.01-Help-And-Documentation.ipynb)에서 설명)을 잘 사용하고 자주 사용하면 유용할 수 있습니다.
여기와 다른 곳의 예제를 살펴보면 I파이썬(Python)이 제공하는 모든 도구에 익숙해지는 데 사용할 수 있습니다.
